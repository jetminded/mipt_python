

## HTML: Основы

### Структура HTML-документа
```html
<!DOCTYPE html>
<html>
  <head>
    <!-- Метаданные и служебная информация -->
    <title>Заголовок страницы</title>
    <meta charset="utf-8">
  </head>
  <body>
    <!-- Основное содержимое страницы -->
  </body>
</html>
```

В HTML всё представлено аттрибутами, как в XML. 
Есть стандартный набор тегов и аттрибутов для них. Нестандартный тег не будет распознан

Самый частый тег **div** (от английского division — раздел, секция) — это блочный HTML-тег, предназначенный для группировки других элементов и структурирования веб-страницы. Сам по себе он не имеет визуального представления, но служит "контейнером" для организации контента и применения стилей.

Есть теги, которые нуждаются в закрывающем теге. Они содержат контент между открывающим и закрывающим тегом.

**Контейнерные элементы:**
```html
<div>Содержимое блока</div>
<p>Текст параграфа</p>
<span>Встроенный текст</span>
<h1>Заголовок</h1>
<a href="#">Текст ссылки</a>
<button>Текст кнопки</button>
<ul><li>Элемент списка</li></ul>
```

**Семантические теги:**
```html
<header>Шапка сайта</header>
<main>Основное содержимое</main>
<footer>Подвал сайта</footer>
<article>Статья</article>
<section>Раздел</section>
<nav>Навигация</nav>
```

**Самозакрывающиеся теги:**

| Тег | Назначение | Пример |
|-----|------------|--------|
| `<img>` | Изображение | `<img src="photo.jpg" alt="Описание">` |
| `<br>` | Перенос строки | `Текст<br>новая строка` |
| `<hr>` | Горизонтальная линия | `<hr>` |
| `<input>` | Поле ввода | `<input type="text" placeholder="Имя">` |
| `<meta>` | Мета-информация | `<meta charset="UTF-8">` |
| `<link>` | Подключение ресурсов | `<link rel="stylesheet" href="style.css">` |
| `<area>` | Область на карте изображения | `<area shape="rect" coords="0,0,100,100" href="page.html">` |
| `<base>` | Базовый URL | `<base href="https://example.com/">` |
| `<col>` | Колонка таблицы | `<col width="100">` |


У тегов есть **атрибуты** (`class`, `id`, `src`, `alt` и пр.). Это дополнительные настройки элемента, которые задаются в открывающем теге. Они предоставляют дополнительную информацию об элементе и влияют на его поведение или отображение.

***

**Основные атрибуты:**
1. `id` - Уникальный идентификатор
	- Задает элементу уникальное имя (на странице не должно быть двух элементов с одинаковым `id`)
	- Используется для стилизации (CSS) и доступа через JavaScript
```html
<div id="header">Шапка сайта</div>
```

2. `class` - Класс элемента
	- Задает один или несколько классов для элемента
	- Классы используются для стилизации (CSS) и выбора элементов в JavaScript
	- Один класс может применяться к множеству элементов
```html
<p class="text bold">Этот текст жирный</p>
<p class="text">А этот нет</p>
```

3. `style` - Встроенные стили
	- Позволяет задавать CSS-стили непосредственно для элемента
```html
<p style="color: red; font-size: 20px;">Красный текст</p>
```

4. `src` - Источник (для медиа-элементов)
	- Указывает путь к файлу (изображение, видео, скрипт)
```html
<img src="images/photo.jpg">
<script src="js/app.js"></script>
```

5. `alt` - Альтернативный текст
	- Текст, который отображается, если изображение не загрузилось
	- Важен для доступности (скринридеры читают этот текст)
```html
<img src="photo.jpg" alt="Описание изображения">
```

6. `href` - Ссылка (для тега `<a>`)
	- Задает URL-адрес, на который ведет ссылка
```html
<a href="https://example.com">Перейти на сайт</a>
```

7. `type` - Тип (для `<input>` и `<button>`)
	- Определяет тип элемента (например, тип поля ввода)
```html
<input type="text">
<input type="password">
<input type="checkbox">
<button type="submit">Отправить</button>
```

8. `placeholder` - Подсказка в поле ввода
	- Текст-подсказка, который исчезает при вводе
```html
<input type="text" placeholder="Введите ваше имя">
```

9. `disabled` - Отключение элемента
	- Делает элемент недоступным для взаимодействия
```html
<button disabled>Неактивная кнопка</button>
```

10. `data-*` - Пользовательские атрибуты
	- Позволяют хранить дополнительную информацию в элементе
	- Используются в JavaScript
```
<div data-user-id="12345" data-role="admin">Панель администратора</div>
```

## CSS: Основы стилизации

### Способы подключения CSS
1. Встроенные стили:
```html
<div style="color: red;"></div>
```

2. Внутри `<head>`:
```html
<style>
  p { color: blue; }
</style>
```

3. Внешний файл (рекомендуемый способ):
```html
<link rel="stylesheet" href="styles.css">
```

### Основные селекторы
1. По тегу: `p { color: red; }`
2. По классу: `.my-class { ... }`
3. По id: `#element-id { ... }`
4. Комбинированные:
   - Потомки: `div p { ... }`
   - Дочерние: `div > p { ... }`
   - Соседние: `h1 + p { ... }`

### Блоковая модель CSS
```
[ Margin (внешний отступ) ]
  [ Border (граница) ]
    [ Padding (внутренний отступ) ]
      [ Content (контент) ]
```

Пример:
```css
.box {
  margin: 10px;
  padding: 20px;
  border: 1px solid black;
  width: 100px;
  height: 100px;
}
```

### Позиционирование
1. `static` (по умолчанию)
2. `relative` - относительно нормального положения
3. `absolute` - относительно ближайшего позиционированного родителя
4. `fixed` - относительно окна браузера
5. `sticky` - "липкое" позиционирование

Пример:
```css
.element {
  position: absolute;
  top: 10px;
  left: 20px;
}
```

### Flexbox
Основные свойства:
```css
.container {
  display: flex;
  flex-direction: row | column;
  justify-content: center | space-between | etc.;
  align-items: center | flex-start | etc.;
  gap: 10px;
}
```

### Grid
Пример сетки:
```css
.container {
  display: grid;
  grid-template-columns: 1fr 1fr 1fr;
  grid-gap: 20px;
}
```

### Адаптивный дизайн
Медиа-запросы:
```css
@media (max-width: 600px) {
  .container {
    grid-template-columns: 1fr;
  }
}
```

# Работа с HTTP и библиотекой `requests` в Python

## Что такое HTTP?

**HTTP (HyperText Transfer Protocol)** — это протокол прикладного уровня, по которому браузеры и приложения общаются с веб-серверами.

Основные понятия:
- **Запрос (request)** — сообщение от клиента к серверу (`GET`, `POST`, `PUT`, `DELETE` и др.)
- **Ответ (response)** — сообщение от сервера: статус-код, заголовки, тело ответа
- **URL** — адрес ресурса, к которому мы обращаемся

### Популярные HTTP-методы

| Метод | Назначение |
|-------|-----------|
| `GET` | Получить данные |
| `POST` | Отправить данные (создать ресурс) |
| `PUT` | Обновить ресурс целиком |
| `PATCH` | Частично обновить ресурс |
| `DELETE` | Удалить ресурс |

### Коды ответов (Status Codes)

| Диапазон | Значение |
|----------|----------|
| `1xx` | Информационные |
| `2xx` | Успех (`200 OK`, `201 Created`) |
| `3xx` | Перенаправление |
| `4xx` | Ошибка клиента (`404 Not Found`, `403 Forbidden`) |
| `5xx` | Ошибка сервера (`500 Internal Server Error`) |

## 1. GET-запрос — получаем данные

Будем использовать открытый тестовый API — [JSONPlaceholder](https://jsonplaceholder.typicode.com).

`GET /posts/1` — получить пост с `id=1`.


````python
# Простейший GET-запрос
response = requests.get("https://jsonplaceholder.typicode.com/posts/1")

# Статус-код ответа
print("Status code:", response.status_code)

# Заголовки ответа (объект, похожий на словарь)
print("\nContent-Type:", response.headers["Content-Type"])

# Тело ответа в виде строки (первые 200 символов)
print("\nRaw text (начало):", response.text)
````

````python
# Если сервер вернул JSON, удобнее распарсить его сразу в dict / list
data = response.json()

print(type(data))  # <class 'dict'>
data
````

### Query-параметры

Часто нужно передать параметры в URL, например `?userId=1&_limit=3`.  
Вместо ручного форматирования строки используйте аргумент `params`:


````python
# Получим первые 3 поста пользователя с userId=1
params = {
    "userId": 1,
    "_limit": 3
}

response = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    params=params
)

# requests сам подставит параметры в URL
print("Фактический URL:", response.url)
print("Количество постов:", len(response.json()))

for post in response.json():
    print(f"  [{post['id']}] {post['title'][:50]}")
````

---
## 2. POST-запрос — отправляем данные

`POST /posts` — создать новый пост.  
Данные передаём через аргумент `json=` (requests автоматически сериализует словарь и выставит `Content-Type: application/json`).


````python
new_post = {
    "title": "Мой первый пост",
    "body": "Привет из Jupyter Notebook!",
    "userId": 42
}

response = requests.post(
    "https://jsonplaceholder.typicode.com/posts",
    json=new_post   # автоматически -> JSON + заголовок Content-Type
)

print("Status code:", response.status_code)   # 201 Created
print("Ответ сервера:", response.json())
````

> **Примечание:** JSONPlaceholder — тестовый сервер. Он *имитирует* создание ресурса и возвращает `201`, но на самом деле ничего не сохраняет.

### Разница между `json=` и `data=`

| Аргумент | Что делает |
|----------|-----------|
| `json={"key": "value"}` | Сериализует в JSON, ставит `Content-Type: application/json` |
| `data={"key": "value"}` | Кодирует как form-data (`application/x-www-form-urlencoded`) |
| `data="raw string"` | Отправляет строку как есть |

В requests это делает параметр data=:

```python
# Отправится как: username=alice&age=25
requests.post(url, data={"username": "alice", "age": "25"})
```

Vs. json=:

```python
# Отправится как: {"username": "alice", "age": "25"}
requests.post(url, json={"username": "alice", "age": "25"})
```

````python
# PUT — полное обновление ресурса
updated_post = {
    "id": 1,
    "title": "Обновлённый заголовок",
    "body": "Новое содержание поста.",
    "userId": 1
}

resp_put = requests.put(
    "https://jsonplaceholder.typicode.com/posts/1",
    json=updated_post
)
print("PUT status:", resp_put.status_code)
print("PUT response:", resp_put.json())
````

````python
# DELETE — удаление ресурса
resp_del = requests.delete("https://jsonplaceholder.typicode.com/posts/1")
print("DELETE status:", resp_del.status_code)   # 200
print("DELETE body:", resp_del.json())           # {} — пустой объект
````

## 4. Заголовки запроса (Headers)

Иногда API требует передавать токены, ключи или другие заголовки.


````python
headers = {
    "Authorization": "Bearer my-secret-token",
    "Accept": "application/json",
    "X-Custom-Header": "hello-seminar"
}

response = requests.get(
    "https://jsonplaceholder.typicode.com/posts/1",
    headers=headers
)

# Посмотрим, какие заголовки мы отправили
print("Заголовки запроса:")
for key, value in response.request.headers.items():
    print(f"  {key}: {value}")
````

## 5. Таймауты и обработка ошибок

В production-коде **всегда** указывайте таймаут, иначе запрос может «зависнуть» навсегда.


````python
# timeout=(connect_timeout, read_timeout) в секундах
try:
    response = requests.get(
        "https://jsonplaceholder.typicode.com/posts/1",
        timeout=(3, 5)   # 3 сек на подключение, 5 сек на чтение
    )
    response.raise_for_status()  # выбросит исключение, если статус 4xx/5xx
    print("Всё ок!", response.status_code)

except requests.exceptions.Timeout:
    print("Сервер не ответил вовремя!")

except requests.exceptions.HTTPError as e:
    print(f"HTTP ошибка: {e}")

except requests.exceptions.ConnectionError:
    print("Не удалось подключиться к серверу.")

except requests.exceptions.RequestException as e:
    print(f"Что-то пошло не так: {e}")
````

## 6. Сессии (`requests.Session`)

Если вы делаете несколько запросов к одному серверу, используйте **сессию**:
- переиспользуется TCP-соединение (быстрее!)
- можно задать общие заголовки / куки / параметры


````python
with requests.Session() as session:
    # Общие настройки для всех запросов в сессии
    session.headers.update({
        "Accept": "application/json",
        "X-Session-Demo": "true"
    })

    # Первый запрос
    r1 = session.get("https://jsonplaceholder.typicode.com/posts/1")
    print("Post 1:", r1.json()["title"][:40])

    # Второй запрос — соединение переиспользуется
    r2 = session.get("https://jsonplaceholder.typicode.com/posts/2")
    print("Post 2:", r2.json()["title"][:40])

    # Заголовок X-Session-Demo был отправлен в обоих запросах
    print("\nHeaders в r2:")
    for k, v in r2.request.headers.items():
        print(f"  {k}: {v}")
````

## 7. Скачивание файлов (бинарный контент)

`requests` умеет работать не только с JSON, но и с любым контентом — картинками, PDF и т.д.


````python
# Скачаем небольшую картинку
image_url = "https://via.placeholder.com/150"

response = requests.get(image_url, timeout=10, )
print("Content-Type:", response.headers.get("Content-Type"), verify=False)
print("Размер:", len(response.content), "байт")

# Сохраним на диск
with open("sample_image.png", "wb") as f:
    f.write(response.content)

print("Файл сохранён: sample_image.png")
````
