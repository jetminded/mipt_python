# Part 1. Compiled and interpreted languages

## 1.1 How does a computer see our code?

A processor does not understand Python. It does not understand C++ or Java
either. A processor understands only **machine code** — long sequences of
numbers, where every number means one very small operation: "copy these bytes",
"add these two registers", "jump to that address".

Our source code, on the other hand, is **text**. Somebody has to translate the
text into machine code. There are two main ways to do this translation:

* **compilation** — translate the whole program *before* running it;
* **interpretation** — analyse and execute the source code piece by piece,
  *while* the program runs.

A program that does the first job is called a **compiler**. A program that does
the second job is called an **interpreter**.

## 1.2 Compilation

**Compilation** translates our code completely into an **executable file**. We
can then run that file on the computer and get the result. The compiler is not
needed any more: it has already done its work.

Classic compiled languages are C, C++, Rust, Go and Fortran.

### The stages of compilation

A compiler does not work in one step. It works in a pipeline, and every stage
solves one problem:

1. **Lexical analysis.** The source code is split into **tokens** — the smallest
   meaningful pieces: names, numbers, operators, brackets. Spaces and comments
   are removed here.
2. **Syntax analysis (parsing).** The compiler checks whether the structure of
   the code is correct for the grammar of the language. A missing bracket is
   found at this stage.
3. **Semantic analysis.** The compiler checks whether variables, data types and
   other elements are used correctly. For example, it will refuse to add a
   number to a text string.
4. **Generating an intermediate representation (IR).** The program is rewritten
   in an internal language of the compiler (for example, LLVM IR). This form is
   easy for a machine to analyse, and it does not depend on any processor.
5. **Optimisation.** This is where the "magic" happens. The compiler makes the
   program faster or smaller: it removes code that can never run, moves
   calculations out of loops, and replaces slow operations with fast ones.
6. **Machine code generation.** Finally the intermediate representation is
   turned into real instructions for the target processor.

### Advantages of compilation

* **Speed.** Compiled programs usually run much faster, because the code is
  already machine code. Nothing has to be translated during the run.
* **Optimisation.** The compiler sees the *whole* program at once, so it can
  apply deep optimisations.
* **No translator at run time.** The user needs only the executable file.
* **Errors are found early.** Syntax and type errors are reported during
  compilation, so they cannot appear later, in front of the user.
* **The source code stays private.** You ship the executable file, not the text.
  This matters for commercial software.
* **Harder to modify.** A compiled program is very difficult to change without
  the source code.

### Disadvantages of compilation

* **Compilation takes time.** The whole program is translated at once. Building
  the Firefox browser takes from several minutes to several hours, depending on
  the computer.
* **Compilation needs a lot of memory.** Firefox officially needs at least
  **4 GB of RAM** just to be built on Windows.
* **You cannot change the program without the source code.** Even a one-line
  change means a full rebuild.
* **One platform at a time.** Code compiled for Windows will not run on Linux,
  and code compiled for an ARM processor will not run on an x86 processor. You
  must compile again for every platform.
* **Debugging is less comfortable.** The program runs as one solid block, so it
  is sometimes hard to see where the problem started.
* **No interactive work.** You cannot "try one line and look at the result",
  because the compiler needs the complete program.

## 1.3 Interpretation

**Interpretation** is the process of executing the source code step by step,
"on the fly". No separate executable file is created.

The stages look shorter than for a compiler:

1. **Lexical analysis** — the source code is split into tokens, and unnecessary
   characters are removed.
2. **Syntax analysis** — the structure of the program is checked against the
   grammar of the language.
3. **Generating an intermediate representation** *(optional)* — some
   interpreters build an internal form of the code to work faster. Python does
   this, and we will look at it in a moment.
4. **Execution** — the interpreter performs the instructions as it reads them.
5. **Debugging** — because the interpreter is present all the time, we can stop
   the program and look at the current values of the variables. This is very
   useful when we are looking for a problem.

### Advantages of interpretation

* **Fast development.** We see the result immediately after we press "run".
  There is no compilation step to wait for.
* **Interactive work.** We can type commands into a console one by one and watch
  what happens. This is exactly what we do in Jupyter.
* **Dynamic typing is easy.** The interpreter knows the type of every value at
  run time anyway, so it can work with variables very flexibly.
* **Little memory.** Only one instruction is translated at a time.
* **Portability.** The same source code runs on every machine that has the
  interpreter. We do not compile it again for each platform.

### Disadvantages of interpretation

* **Low performance.** Every instruction is processed separately, again and
  again. A loop body is re-interpreted on every iteration.
* **The interpreter must be installed.** To run the program, the user needs a
  suitable interpreter on their computer.
* **Almost no optimisation.** The interpreter looks at a small piece of the
  program at a time, so it cannot do the whole-program analysis that a compiler
  does.
* **Some errors appear only at run time.** A misspelled name in a branch that is
  rarely executed can survive for months.

## 1.4 What CPython really does: bytecode

Now the important detail. **CPython** — the standard Python interpreter, written
in the C language — does *not* execute our source text directly.

First it translates the source code into **bytecode**. Bytecode is a compact set
of instructions, but they are *not* instructions for a real processor. They are
instructions for a **virtual machine** — a program inside CPython that imitates
a simple processor. The virtual machine is basically one big loop: it takes the
next bytecode instruction and performs it.

So the honest description of Python is:

> Python source code is **compiled into bytecode**, and the bytecode is then
> **interpreted** by a virtual machine.

You have already seen the result of this without knowing it. When you import
your own module, Python saves the bytecode next to it in a folder called
`__pycache__`, in a file with the extension `.pyc`. Next time the module has not
changed, Python simply loads the ready bytecode and skips the translation.

**Why use bytecode at all?** Parsing text is slow, and in a loop we would have to
parse the same text thousands of times. Bytecode is parsed once. It is also
compact and portable: the same bytecode works on Windows, Linux and macOS.

**What is the price?** The virtual machine stands between our program and the
real processor. It hides the processor, so the deep optimisations of a C compiler
are impossible. This layer is one of the main reasons why Python is slower than
C. In exchange we get all the dynamic flexibility of the language — and, as we
will see in Part 2, that flexibility is not free either.

## 1.5 Compiled and interpreted side by side

| | Compiled languages | Interpreted languages |
|---|---|---|
| **Is the translator needed to run the program?** | No, only during compilation | Yes, every single time |
| **How much can be optimised?** | A lot — the whole program is analysed | Little — the code is processed piece by piece |
| **Interactive work** | Difficult: the compiler needs the complete code | Natural: commands can be entered live |
| **Moving to another platform** | Compile again for each platform | The same code (or bytecode) runs anywhere |
| **Speed of the program** | High | Lower |
| **Speed of development** | Lower | High |
| **Examples** | C, C++, Rust, Go | Python, JavaScript, Ruby, PHP |

### The border is not sharp

Real languages do not fit perfectly into two boxes.

* **Java and C#** compile to bytecode, like Python, and then a virtual machine
  runs it.
* **PyPy** is another Python implementation. It uses **JIT compilation**
  (*just-in-time*): while the program runs, PyPy notices which parts are
  executed most often — the "hot" code — and compiles exactly those parts into
  real machine code. On heavy calculations it can be several times faster than
  CPython.
* **Numba** does the same trick inside CPython for chosen functions, which is
  very popular in scientific code.
* **Cython** goes the other way: it translates Python-like code into C, and then
  a C compiler builds it.

So "Python is an interpreted language" is a statement about the usual
implementation, not about some eternal law.

## 1.6 Typing: how a language treats types

Compilation and interpretation answer the question *when* our code is
translated. **Typing** answers a different question: how the language
**recognises the types** of our values and what it allows us to do with them.

There are three independent pairs of terms. A language has one property from
each pair.

### Strong or weak typing

This pair is about mixing types in one operation. Look at the expression:

```python
1 + '1'
```

* In a **strongly typed** language this is an error. The language refuses to
  guess whether we want the number `2` or the text `'11'`. Examples: **Python**,
  Java, Ruby.
* In a **weakly typed** language the value is converted automatically, following
  the rules of the language. In JavaScript `1 + '1'` gives the string `'11'`.
  Examples: **JavaScript**, C, C++, PHP, Perl.

Strong typing finds mistakes earlier and makes the behaviour of the code
predictable. Weak typing is comfortable for quick prototypes, but it can produce
surprising results.

We met this rule in the first lesson: `'2' - 1` raises a `TypeError`, and we have
to write `int('2') - 1` if we want a number.

### Static or dynamic typing

This pair is about the *moment* when the type is fixed and checked.

* In a **statically typed** language the type of a variable is known **before**
  the program runs, and the compiler checks it during compilation. Many mistakes
  are caught before the program starts. Examples: C++, Java, C#.
* In a **dynamically typed** language the type is determined **while the program
  runs**. The same name can hold a number now and a string one line later. This
  is flexible, but a type mistake becomes a run-time error. Examples:
  **Python**, JavaScript.

A small C++ example of static typing:

```cpp
int x = 5;        // x is an integer, for ever
x = "Hello";      // error at compile time: cannot assign const char* to int
```

````python
x = 5
x = "Hello"
````

### Explicit or implicit typing

This pair is about *who writes* the type.

* In a language with **explicit typing** we must write the type ourselves when we
  declare a variable: `int x = 5;`. Examples: C++, C#, Rust.
* In a language with **implicit typing** we do not write the type anywhere. It is
  determined by the value we assign. Examples: **Python**, PHP, JavaScript.

### Where is Python?

> Python is a language with **strong**, **dynamic**, **implicit** typing.

This combination is exactly why Part 2 of this lesson exists.

Think about it. The name `x` carries no type — that is implicit typing. Nothing
is checked before the program runs — that is dynamic typing. And yet Python
always knows the exact type of every value and always refuses to mix types by
accident — that is strong typing.

If the *name* does not carry the type, then the type must be stored somewhere
else. It is stored **inside the value itself**. Every value in Python is an
object that carries its own type with it. Let us see what such an object looks
like in memory.

## 1.7 A practical note: two ways of working with Python

Because Python is interpreted, we can work with it in two different styles, and
both are useful.

**"Write and run".** We write the whole program in an editor, save it in a file
with the extension `.py`, and then run it from the command line:

```bash
python my_program.py
```

This style is right for large projects, for desktop and server applications, and
for debugging with a debugger.

**Interactive.** We execute the code piece by piece and look at the result of
each piece before we write the next one. The simplest way is to start the
interpreter with no file name:

```bash
python
```

This style is right for data analysis (we want to see the table or the graph and
then decide what to do next), for quick experiments, and for reports where code,
graphs and comments live in one document.

A plain console is not comfortable for writing, so we use better tools:
**Jupyter Notebook** and **JupyterLab** (the classic choice — this lesson is a
Jupyter notebook), **VS Code** with the Jupyter extension, or a full IDE such as
**PyCharm**.

# Part 2. Objects inside Python: `PyObject`, hashing, dictionaries

## 2.1 Everything in Python is an object

In Python there are no "simple" values that live outside the object system. A
number is an object. A string is an object. A list is an object. A function is an
object, and even a type is an object.

This is not a philosophical statement. It is a description of the memory. Every
value that we create lives in memory as a structure with a fixed beginning, and
that beginning is the same for all of them.

## 2.2 `PyObject` — the head of every object

PyObject is a fundamental structure in CPython (the Python interpreter written in C) that serves as the foundation for all objects in the language. CPython is written in C. In C, a structure describes how bytes are arranged in
memory. Every Python object begins with this structure:

```c
typedef struct {
    Py_ssize_t    ob_refcnt;   /* reference count           */
    PyTypeObject *ob_type;     /* pointer to the type object */
} PyObject;
```

Two fields only, and both of them explain something we can observe from Python:

* **`ob_type`** — a pointer to the **type** of the object. This is the field that
  the function `type()` reads. Here is the answer to the question from Part 1:
  the type is not stored in the variable, it is stored in the object. This is how
  a language with implicit, dynamic typing can still be strongly typed. When
  Python evaluates `a + b`, it looks at `ob_type` of both objects and asks the
  types how to add them. If the types do not agree, we get a `TypeError`.
* **`ob_refcnt`** — the **reference count**: how many names, list items,
  dictionary values and other places currently point to this object.

After these two fields comes the data of the concrete type: the digits of the
number, the characters of the string, the pointers of the list.

### What the reference count is for

The reference counting mechanism is used to determine which objects can be deleted.

* when a new name starts to point to an object, `ob_refcnt` grows by one;
* when a name stops pointing to it (we reassign it, or the function ends), the
  count falls by one;
* when the count reaches **zero**, nobody can reach the object any more, so its
  memory is freed immediately.

Reference counting alone cannot free a **cycle** — for example, a list that
contains itself. Its count never reaches zero. For this case CPython also has a
second, slower mechanism, the *garbage collector*, which searches for such
groups of objects and removes them.

Let us look at the object head from Python.

````python
import sys

x = 'hello'

print('type:       ', type(x))          # this reads ob_type
print('identity:   ', id(x))            # the address of the object in memory
print('size, bytes:', sys.getsizeof(x))
````

The function `id()` returns the identity of the object — in CPython, its address
in memory. We will use it a lot in a minute.

Notice the size: a short string of five characters takes far more than five
bytes. Those extra bytes are the object head plus the service fields of the
string type. This is the price of "everything is an object": Python objects are
much bigger than the raw data inside them. A C program storing the same five
characters needs six bytes.

````python
from sys import getrefcount

a = []
print(getrefcount(a))
````

````python
b = a
print(getrefcount(a))
c = a
print(getrefcount(a))
````

## 2.3 `PyVarObject` — objects of variable length

Some objects always have the same size. A `float` always holds exactly one
floating-point number. `None` holds nothing at all. For these objects the two
fields of `PyObject` plus the data are enough.

But many objects hold a **different number of items** each time. A list can
have three items or three thousand. A string can be one character or a whole
book. Such objects need one
more field, and CPython describes them with an extended structure:

```c
typedef struct {
    PyObject   ob_base;   /* refcount + type      */
    Py_ssize_t ob_size;   /* the number of items  */
} PyVarObject;
```

`PyVarObject` is a `PyObject` **plus** `ob_size`, the number of items the object
currently holds. This is the field that `len()` reads for a tuple or a string.



````python
import sys

print(sys.getsizeof(()))            # empty tuple: only the head
print(sys.getsizeof((1, 2, 3)))     # three items
print(sys.getsizeof(tuple(range(10))))  # ten items

print(len((1, 2, 3)))               # len() reads ob_size
````

## 2.4 A variable is a name, not a box

Many textbooks say that a variable is a box where we put a value. For Python it
is more correct to say:

> A variable is a **name** (a label) attached to an object in memory.

Assignment **never copies** the object. It only makes a second label for the same
object. The function `id()` lets us prove it.

````python
list_1 = [1, 2, 3]
list_2 = list_1          # not a copy: a second name for the same list

print(id(list_1), id(list_2), id(list_1) == id(list_2))

list_2.append(4)
print('list_1:', list_1)   # the "first" list changed too
print('list_2:', list_2)
````

Nothing strange happened here: there was only ever **one** list, with two names.
If we want an independent object, we must ask for a copy.

````python
list_1 = [1, 2, 3]
list_2 = list_1.copy()   # now this is a real, separate list

list_2.append(4)
print('list_1:', list_1)
print('list_2:', list_2)
print(id(list_1) == id(list_2))
````

With numbers the same trick does not happen:

````python
g = 10
k = g
k += 1        # k = k + 1 creates a NEW object and points k at it

print('g =', g, ' k =', k)
````

The reason is not that numbers are special in some magic way. The reason is that
numbers are **immutable**, and this is the next idea.

### `is` and `==` answer two different questions

* `==` compares **values**: "do these two objects look the same?"
* `is` compares **identity**: "are these two names pointing to one and the same
  object?" It is the same as comparing `id()`.

For small integers the result of `is` can look surprising, because CPython
creates the objects for the numbers from `-5` to `256` once, at start-up, and
reuses them. Short strings are cached in a similar way.

````python
a = 256
b = 256
print(a == b, a is b)          # a cached number: one object, two names

x = 5000
y = 5000               # this number is built while the program runs
print(x == y, x is y)          # equal values, but two different objects
````



This is a good illustration of the whole idea: *when* two equal values are the
same object is an implementation detail of CPython. It can differ between
versions and between situations, and we must never build our logic on it.

> **Practical rule.** Use `==` when you compare values. Use `is` only with
> `None` (`if x is None`), and with `True` / `False` if you really mean identity.

## 2.5 Mutable and immutable objects

An object is **immutable** if its contents can never change during its whole
life. We can point a name at a different object, but we cannot edit the object
itself.

| | Types |
|---|---|
| **Immutable** | `int`, `float`, `bool`, `complex`, `str`, `bytes`, `tuple`, `frozenset`, `NoneType` |
| **Mutable** | `list`, `dict`, `set` |

This one property explains the difference between the list example and the number
example above. `list.append` changes the object, so every name sees the change.
`k += 1` cannot change the number `10`, so Python builds `11` as a new object and
moves only the name `k`.

Immutability is not only a rule about what we may write. It is a **guarantee**,
and in the next section we build something very useful on top of it.

## 2.6 Hash functions

A **hash function** is a function that takes input data of *any* length and,
following a fixed algorithm, returns a value of a *fixed* size. That value is
called the **hash** (or hash code) of the data.

A hash function has three properties that interest us:

1. **It is stable.** The same input always gives the same hash.
2. **It is sensitive.** If we change even one letter of the input, the hash
   changes completely — not "a little", but beyond recognition.
3. **It is irreversible (one-way).** We cannot take the algorithm backwards and
   read the original message from the hash. If an attacker steals a database of
   *hashed* passwords, the only way forward is to try candidate passwords one by
   one, and that can take months or years.

There is a fourth property that we will need in a moment: computing a hash is
**fast**, and it does not depend on how much data we already store. Finding
something by its hash code takes much less time than looking through everything
one item at a time.

### Where hash functions are used

* **Passwords and digital signatures.** A serious service never stores your
  password; it stores its hash and compares hashes.
* **Checking that a file is not damaged or changed.** A game can check that the
  player has not edited the game files. A download page publishes the hash of the
  archive so that you can verify your copy.
* **Version control systems.** Git names every object — every file, commit and
  tree — by the hash of its contents, and uses this to check that the repository
  is intact.
* **Blockchain.** Every block contains the hash of the previous block. If someone
  changes an old transaction, all the following hashes stop matching, and the
  network rejects the change.
* **Hash tables** — the data structure behind Python's `dict` and `set`. This is
  our subject today.

In Python we can compute a hash with the built-in function `hash()`. It always
returns an `int`.

````python
print(hash('hello'))
print(type(hash('hello')))

print(hash(343))
print(hash(-343))
print(hash(True))
print(hash((1, 2, 3)))
````

## 2.7 Hashability in Python

Not every object can be hashed. The official definition from the Python
documentation ([link](https://docs.python.org/3/glossary.html#term-hashable)) is:

> An object is **hashable** if it has a hash value which **never changes during
> its lifetime** (it needs a `__hash__()` method), and can be compared to other
> objects (it needs an `__eq__()` method). Hashable objects which compare equal
> must have the same hash value.

Read the first condition again, because it is the whole point: the hash must
never change. And here immutability comes back.

> **The rule is simple: immutable objects are hashable, mutable objects are not.**

| | Types |
|---|---|
| **Hashable** | `str`, `int`, `bool`, `float`, `bytes`, `tuple` *(with conditions)*, `frozenset`, `None`, functions, classes |
| **Not hashable** | `list`, `dict`, `set` |

A list is not hashable *by design*. If a list had a hash, the hash would change
every time we appended an item — and that breaks the contract above.

We can check hashability without triggering an error, by asking whether the type
has a real `__hash__` method:

````python
print('List: ', [1, 2].__hash__ is not None)
print('Tuple:', (1, 2).__hash__ is not None)
print('Dict: ', {1: 2}.__hash__ is not None)
print('Str:  ', 'abc'.__hash__ is not None)
````

### The hashability of tuples

**Careful:** not every tuple is hashable! A tuple is hashable only if **all of
its items** are hashable. This makes sense: the hash of the tuple is computed
from the hashes of its items, so a mutable item would make the tuple's hash
unstable.

````python
print(hash((1, 2, (3, 4))))    # a tuple inside a tuple: fine
print(hash((1, 2, [3, 4])))    # a list inside a tuple: TypeError
````

### Hashing `float` — be careful

A `float` is hashable, but using floats as dictionary keys is usually a bad idea.
The reason is: a `float` is stored in binary
form and is not exact. Two numbers that look different in our text may be the
*same* object after rounding, and two numbers that look almost equal may have
completely different hashes.

````python
hash1 = hash(6.5)
hash2 = hash(6.500000000000001)
hash3 = hash(6.5000000000000001)   # too many digits: this IS 6.5 for a float
hash4 = hash(round(6.50443, 2))

print('hash 1:', hash1)
print('hash 2:', hash2)
print('hash 3:', hash3)
print('hash 4:', hash4)
````

`hash3` is equal to `hash1`, because the literal `6.5000000000000001` cannot be
represented and becomes exactly `6.5`. But `hash2` is different. If we use such
values as keys, we will look for a key and not find it.

> If you need exact decimal keys (money, for example), use the `decimal` module
> or store the value as an `int` of cents.

### Two more notes on `hash()`

**Note 1.** For strings and bytes, the hash value **changes between runs of the
interpreter**. Try `hash('hello')` now, restart the kernel, and try again — you
will get a different number. This is deliberate protection against attacks in
which someone sends specially chosen keys to make a server's dictionaries slow.
The behaviour is controlled by the environment variable `PYTHONHASHSEED`. Hashes
of numbers do *not* change.

So: never save a hash value to a file and never send it to another process
expecting it to match. Inside one run it is perfectly stable, and that is all a
dictionary needs.

**Note 2.** Equal objects must have equal hashes — and in Python `1 == 1.0` is
`True`. So all these numbers share one hash value:

````python
print(hash(1), hash(1.0), hash(True))

s = {1, 0, 0.0, False, True}     # five items written, how many are stored?
print(s)
````

Only two items survive. `1`, `1.0` and `True` are *equal*, so for a set they are
one and the same key, and the first one written wins. This is a favourite trap in
interviews, and now you know exactly why it happens.

## 2.8 How a hash table works

We are ready for the main idea of this part. A `dict` is a **hash table**, and
this is how a hash table stores a pair *key → value*:

1. There is an array of **slots** in memory — say, eight of them at the start.
2. To store `d[key] = value`, Python computes `hash(key)` and turns that number
   into a slot number (roughly `hash(key) % number_of_slots`).
3. The pair is written into that slot.

And now the beautiful part. To *find* `d[key]`, Python does exactly the same
calculation and goes **straight to the right slot**. It does not look through the
other keys at all.

> This is why looking up a key in a dictionary takes about the **same time** for
> a dictionary of ten items and for a dictionary of ten million items. In the
> language of algorithms, the average cost is **O(1)** — constant. Searching for
> a value in a list means checking the items one by one, which is **O(n)**.

**Collisions.** Two different keys can produce the same slot number. This is
normal and unavoidable, because there are many more possible keys than slots.
When it happens, Python looks at the next slots by a fixed rule until it finds a
free one (when storing) or the right key (when searching). To decide whether it
has found "the right key", it compares the keys with `==`. This is why a hashable
object needs **both** `__hash__` and `__eq__`.

**Growing.** When the table becomes about two thirds full, collisions become
frequent, so Python allocates a bigger array and places all the keys again. This
is why the *order* of items in a `set` can look random: it depends on the hash
values and on the size of the table, not on the order in which we wrote them.

**The price.** A hash table needs free slots to stay fast, so a dictionary uses
noticeably more memory than a list holding the same data. We trade memory for
speed.

### Now everything fits together

Look back at the rule from section 2.7. A key must be hashable — that is, in
practice, immutable — because **the slot is chosen from the hash**. If we could
change a key after putting it in the dictionary, its hash would change, its
correct slot would change, but the pair would still be lying in the old slot.
Python would compute the new slot, look there, find nothing, and report that the
key does not exist. The value would be lost inside our own dictionary.

## 2.9 `dict` — the dictionary

A **dictionary** (`dict`) is a container that stores pairs **key → value**. It is
called a *mapping*: it maps keys onto values.

* The **keys** must be **hashable**, and they are unique.
* The **values** can be absolutely anything, and they may repeat.

In C++ the closest relative is `std::unordered_map`, but a Python dictionary can
mix types freely.

### Creating a dictionary

We can use curly brackets or the function `dict()`:

````python
dict_example = {'ivan': 100, 'peter': 200, -10: 'new', (1, -10): True}

print(dict_example)
print(type(dict_example))
````

Note the keys in this example: a string, a negative number and a tuple. All of
them are hashable, so all of them are allowed.

The function `dict()` accepts a list of pairs:

````python
dict_example = dict([
    ('ivan', 100),
    ('peter', 200),
    (-10, 'new'),
    ((1, -10), True),
])
print(dict_example)
````

…or keyword arguments, if the keys are valid Python names:

````python
dict_example = dict(
    ivan=100,
    peter=200,
    key='new',
)
print(dict_example)
````

The most convenient way to build a dictionary from two lists — a list of keys and
a list of values — is the function `zip()`, which joins them in pairs. We will
meet `zip` again in Part 3.

````python
keys = ['ivan', 'peter', -10, (1, -10)]
values = [100, 200, 'new', True]

dict_example = dict(zip(keys, values))
print(dict_example)
````

### Operations with dictionaries

To reach a value, we put the key in square brackets. Adding, changing and
removing also work through the key.

````python
dict_example = {'ivan': 100, 'peter': 200, -10: 'new', (1, -10): True}

print(dict_example['ivan'])
````

````python
dict_example = {'ivan': 100, 'peter': 200, -10: 'new', (1, -10): True}

# change an existing value
dict_example['ivan'] = 200
print(dict_example)

# add a new pair -- the same syntax
dict_example['ivan_the_great'] = 300
print(dict_example)

# remove a pair
del dict_example['ivan']
print(dict_example)

# check that a key is present
print('peter' in dict_example)
print(dict_example.get('abacaba') is None)
````

Notice that changing a value and adding a new pair look exactly the same. If the
key exists, the value is replaced; if it does not, the pair is created.

Asking for a key that does not exist is an error:

````python
dict_example = {'peter': 200, -10: 'new'}
print(dict_example['ivan_the_greatest'])
````

A `KeyError` is often not what we want. The method `get()` is the safe version:
it returns `None` instead of raising an error, and we can give our own default
value as a second argument.

````python
scores = {'ivan': 100, 'peter': 200}

print(scores.get('ivan'))
print(scores.get('maria'))          # None -- no error
print(scores.get('maria', 0))       # our own default value
````

### Useful methods of a dictionary

| Method | What it does |
|---|---|
| `d.keys()` | a view of all the keys |
| `d.values()` | a view of all the values |
| `d.items()` | a view of all the pairs `(key, value)` |
| `d.get(k)`, `d.get(k, default)` | safe reading |
| `d.pop(k)` | remove the key and return its value |
| `d.update(other)` | add or overwrite pairs from another dictionary |
| `len(d)` | how many pairs there are |

> **Note about order.** Since **Python 3.7** a `dict` keeps the order in which
> the keys were added, and this is a guarantee of the language, not an accident.
> Before 3.7 we had to use a special type, `collections.OrderedDict`, when the
> order mattered. A `set`, as we will see, gives no such guarantee.

## 2.10 `set` — the set

A **set** (`set`) is a container that stores a collection of **unique** items. It
is built on the same hash table as a dictionary — you can think of it as a
dictionary that keeps only the keys and no values. That is why the rules are the
same: the items of a set must be **hashable**.

The basic operations are:

* add an item to the set;
* remove an item from the set;
* check whether an item is in the set.

A set is the right tool when we need:

* the **unique** items of a list;
* a **fast** membership test (`in` on a set is O(1); `in` on a list is O(n));
* the algebra of sets — union, intersection, difference.

### Creating a set

We can use curly brackets or the function `set()`:

````python
set_example = {1, 2, 3}
print(set_example, type(set_example))

set_example = set([2, 3, 4])       # from a list
print(set_example, type(set_example))

print(list(set([1, 1, 2, 3])))     # the classic use: remove duplicates
````

> **Careful:** an empty set can be created **only** as `set()`. The brackets `{}`
> give an empty **dictionary**, because dictionaries got the curly brackets
> first.

````python
x = {}
print(type(x))

x = set()
print(type(x))
````

### Operations with a set

To add, remove and test items we use `add()`, `remove()`, `discard()` and the
operator `in`.

````python
set_example = {1, '2', 3.0, 'four'}
print(set_example)

set_example.add(5)                 # add
print(set_example)

set_example.remove(1)              # remove
print(set_example)

set_example.discard(3)             # 3 == 3.0, so this removes 3.0
print(set_example)

print(1 in set_example)            # test
````

**What is the difference between `remove` and `discard`?** Both remove an item,
but they behave differently when the item is not there: `discard` does nothing,
while `remove` raises a `KeyError`.

````python
set_example = {'2', 'four', 5}

set_example.discard(3)     # 3 is not in the set -- nothing happens
print(set_example)
````

````python
set_example.remove(3)      # 3 is not in the set -- KeyError
````

Because a set stores only **unique** items, adding the same item again changes
nothing at all:

````python
set_example = {'2', 'four', 5}
print(set_example)

set_example.add(1)
set_example.add(1)
set_example.add(1)

print(set_example)
````

### Combining and comparing sets

`update()` adds all the items of another set:

````python
a = {1, 2, 3}
b = set([2, 3, 4])

b.update(a)                 # b becomes the union of b and a
print(a, b)
````

Besides testing one item with `in`, we can ask how two whole sets are related:

````python
a = {1, 2, 3}
b = {1, 2, 3, 4}

print(b.issubset(a))      # is b a subset of a?
print(b.issuperset(a))    # is b a superset of a?
print(b.isdisjoint(a))    # have they no common items?
````

### The algebra of sets

The four classic set operations have short operator forms:

````python
a = {1, 2, 3, 5}
b = set([2, 3, 4, 5, 6])

print(a - b)   # difference:           items of a that are not in b
print(b - a)   # difference:           items of b that are not in a
print(a | b)   # union:                items in a or in b
print(a & b)   # intersection:         items in both a and b
print(a ^ b)   # symmetric difference: items in exactly one of the two
````

The same operations also exist as methods, with readable names. The `..._update`
versions change the set in place, exactly like the `-=` and `|=` operators.

````python
a = {1, 2, 3, 5}
b = {2, 3, 4, 5, 6}

print(a.difference(b))            # a - b
print(a.union(b))                 # a | b
print(a.intersection(b))          # a & b
print(a.symmetric_difference(b))  # a ^ b
````

### `frozenset` — the immutable set

A `set` is mutable, so a `set` is **not hashable** and cannot be a dictionary key
or an item of another set. For those cases Python has **`frozenset`**: it is
created once and can never be changed, which makes it hashable.

````python
key = frozenset({'a', 'b'})
print(hash(key) is not None)

groups = {frozenset({'a', 'b'}): 'pair AB'}   # a set as a dictionary key
print(groups[frozenset({'b', 'a'})])          # the order does not matter
````

> There is no `frozendict` in the standard library, although it is often asked
> for. If you need a read-only mapping, look at
> `types.MappingProxyType`.

# Part 3. Conditions and loops

Until now our programs were straight lines: every instruction was executed once,
from top to bottom. Real programs need two more powers:

* to **choose** — do this, but only in certain cases;
* to **repeat** — do this many times.

The first is a **condition**, the second is a **loop**.

## 3.1 Blocks and indentation

Before the first `if`, one word about the syntax of Python.

Most languages mark a block of code with curly brackets `{ }`. Python marks it
with **indentation** — the empty space at the beginning of the line. Here
indentation is not decoration, it is **syntax**:

```python
header line:
    the block
    belongs to the header
this line is outside the block again
```

Two rules follow:

1. The header line always ends with a **colon** `:`.
2. Every line of the block has the same indentation. **PEP 8** asks for
   **4 spaces** per level.

If the indentation is wrong, the program does not start: Python raises an
`IndentationError`. And never mix tabs with spaces in one file — it looks the
same on the screen but produces a `TabError`. Any modern editor can insert
spaces when you press *Tab*; turn that option on and forget the problem.

## 3.2 The conditional operator `if`

The `if` statement runs a block **only if** a condition is true:

```python
if condition:
    do_something()
```

The condition is any expression. Python converts its result to `bool`.

````python
temperature = 30

if temperature > 25:
    print('It is hot today.')
    print('Take some water with you.')

print('This line runs in any case.')
````

### `else` and `elif`

`else` gives the alternative — the block that runs when the condition is false:

````python
age = 15

if age >= 18:
    print('You can vote.')
else:
    print('You cannot vote yet.')
````

When there are more than two cases, we add `elif` (short for *else if*). Python
checks the conditions **from top to bottom** and executes the block of the
**first** one that is true. All the others are skipped, even if they are true
too. The final `else` is optional and catches everything that is left.

````python
grade = 4

if grade == 5:
    print('excellent')
elif grade == 4:
    print('good')
elif grade == 3:
    print('satisfactory')
else:
    print('bad')
````

Because only the first true branch runs, the **order matters**. In the example
below the second condition can never be reached: every number that is greater
than 100 is also greater than 10.

````python
n = 500

if n > 10:
    print('greater than 10')
elif n > 100:              # this branch is unreachable!
    print('greater than 100')
````

### What counts as true?

The condition does not have to be a comparison. Python can convert any object to
`bool`, and the rule is short:

**These values are `False`:**

* `False` and `None`;
* zero of any numeric type: `0`, `0.0`, `0j`;
* every **empty** container: `''`, `[]`, `()`, `{}`, `set()`.

**Everything else is `True`** — including the string `'0'` and the list `[0]`,
which are not empty.

This is why Python programmers do not write `if len(my_list) > 0:`. They write
`if my_list:`, which means the same and reads better.

````python
my_list = []

if my_list:
    print('the list has items')
else:
    print('the list is empty')

print(bool(0), bool(''), bool([]), bool(None))
print(bool('0'), bool([0]), bool(-1))
````

### Chained comparisons

In Python we can write a double inequality exactly as in mathematics. This is
shorter and clearer than joining two comparisons with `and`.

````python
x = 7

print(0 < x < 10)          # instead of 0 < x and x < 10
print(1 <= x <= 5)
````

### `and` and `or` are lazy

The operators `and` and `or` use **short-circuit evaluation**: they stop as soon
as the answer is known.

* In `a and b`, if `a` is false, the answer is already false, so `b` is **never
  evaluated**.
* In `a or b`, if `a` is true, the answer is already true, so `b` is **never
  evaluated**.

This is not a small detail — we use it to write safe checks. In the example
below, division by zero can never happen, because the second part is not
evaluated when `n` is `0`.

````python
n = 0

if n != 0 and 10 / n > 1:
    print('the quotient is greater than one')
else:
    print('nothing happened, and there was no error')
````

The same trick protects us from an index error, or from calling a method on
`None`:

```python
if my_list and my_list[0] == 'first':      # safe on an empty list
    ...
if name is not None and name.startswith('A'):
    ...
```

### The conditional expression (the "ternary operator")

Sometimes we only need to **choose a value**, not to run a block. Then we can use
a conditional expression, which fits on one line:

```python
value_if_true if condition else value_if_false
```

````python
x = 10101

print('less than 5' if x < 5 else 'not less than 5')

age = 20
status = 'adult' if age >= 18 else 'child'
print(status)
````

> **Note.** A conditional expression has no `elif`. We can nest them —
> `a if c1 else b if c2 else c` — but this becomes unreadable very quickly. If
> you need more than one condition, use a normal `if` / `elif` / `else`.

### `match` — a modern alternative *(Python 3.10 and later)*

When we compare **one value** against many possible variants, a long chain of
`elif` is tiring to read. Since version 3.10 Python has the `match` statement:

```python
match command:
    case 'start':
        ...
    case 'stop':
        ...
    case _:              # "_" means "anything else", like else
        ...
```

`match` can do much more than compare values — it can also take a structure
apart. We will not use it in this course, but it is useful to recognise it when
you see it in modern code.

## 3.3 Loops

A **loop** lets us perform a set of commands several times. Python has two kinds
of loop:

* **`while`** — repeat *as long as a condition is true*. Use it when we do not
  know in advance how many repetitions there will be.
* **`for`** — repeat *for every item of a collection*. Use it when we do know, or
  when we simply want to visit every item.

One repetition of a loop is called an **iteration**.

## 3.4 The `while` loop

It is often necessary to continue doing something until a condition stops being
true. The syntax is:

```python
while condition:
    do_something()
```

Python checks the condition, runs the block, checks the condition again, and so
on. For example, a loop that prints a greeting three times:

````python
greetings = 0

while greetings < 3:
    print(greetings, 'Hello!')
    greetings = greetings + 1
````

> **The most common mistake with `while`.** The body of the loop **must change
> something that the condition tests**. If we forget the line
> `greetings = greetings + 1`, the condition stays true for ever and the program
> hangs. In Jupyter you stop such a loop with the ⏹ *(interrupt)* button; in a
> terminal, with `Ctrl+C`.

### The `break` statement

We can interrupt a loop at any moment with the `break` statement. It stops the
**whole** loop immediately.

Here we deliberately write `while True:` — a loop with no end — and leave it with
`break`. This is a normal, common pattern when the exit condition is easier to
express in the middle of the body.

````python
i = 0
int_sum = 0

# the sum of the first 100 natural numbers
while True:
    i += 1
    int_sum += i
    if not i < 100:
        break

print(int_sum)
````

> **A note about names.** It would be natural to call the variable `sum`, but
> `sum` is the name of a function from the standard library. Technically Python
> lets us hide `sum` behind our own variable, but it is strongly not recommended:
> the function stops working for the rest of the program, and the error message
> we get much later will be very confusing. Other names not to use as variables:
> `int`, `str`, `list`, `dict`, `set`, `max`, `min`, `len`, `type`, `id`,
> `input`, `next`, `lambda`.

### The `continue` statement

Sometimes we do not want to leave the loop — we only want to **skip the rest of
this iteration** and go straight to the next one. For example, when only some of
the items need to be processed. This is what `continue` does.

````python
i = 0
k = 5

summ = 0  # the sum of the first 100 numbers that are NOT multiples of k

while i < 100:
    i += 1
    if not i % k:       # "not i % k" is the same as "i % k == 0"
        continue

    summ += i

print(summ)
````

**Exercise** 🤗

Can we write this loop without `continue`?

````python
# place for your code
````

Yes, of course. `continue` never adds new power to the language — it only changes
how the code reads. Here the version without it is just as clear:

````python
i = 0
k = 5

summ = 0  # the sum of the first 100 numbers that are NOT multiples of k

while i < 100:
    if i % k != 0:
        summ += i
    i += 1

print(summ)
````

> `continue` is most useful when the body of the loop is long, or when there are
> several reasons to skip an item. Then a few short checks at the top save us
> from a deeply nested `if` at the bottom.

### `else` after a loop

Python has one unusual feature: a loop may have an `else` block. It runs when the
loop finished **normally**, that is, **without** `break`. If the loop was left by
`break`, the `else` block is skipped.

This is exactly what we need for a search: "look through everything; if you did
not find it, say so".

````python
number = 91

divisor = 2
while divisor * divisor <= number:
    if number % divisor == 0:
        print(number, 'is divisible by', divisor)
        break
    divisor += 1
else:
    print(number, 'is a prime number')
````

## 3.5 The `for` loop

In Python the `for` loop is arranged like this:

```python
for variable in iterable:
    commands(variable)
```

Note that this is **not** the `for` of C++ or Java. There is no counter, no
condition and no step. A Python `for` loop says: "take the items of this
collection one by one, put each of them into `variable`, and run the block".

### What can we loop over? Iterable objects

Technically, `iterable` can be any object that is **iterable** — that is, an
object that can give out its items one by one. Formally, an iterable provides the
method `__iter__()`, which returns an *iterator*, and the iterator provides the
method `__next__()`, which returns the next item.

Almost every container in Python is iterable. But iterable objects are divided
into two groups, and the difference is important:

**Sequences** — the order of the items is defined, and we can also take an item
by its index:

* `list`
* `tuple`
* `str`
* `range`

**Iterables that are not sequences** — we can walk through all the items, but
there is no index and (for some of them) no guaranteed order:

* `set` and `frozenset`
* `dict`
* generators
* open files

> **This is the difference to remember.** "Not a sequence" does **not** mean "not
> iterable". A `set` and a `dict` work perfectly well in a `for` loop — that is
> what the rest of this section shows. What they do not support is `x[0]`.

### `for` over sequences

Examples of `for` loops over a list, a tuple and a string:

````python
list_example = ['list', 1, 2, 3, 4, 5]

for item in list_example:   # "item" is a common name for the loop variable
    print(item, end=' ')

print()

tuple_example = ('tuple', 1, 2, 3, 4, 5)
for item in tuple_example:
    print(item, end=' ')

print()

string_example = 'string12345'
for item in string_example:
    print(item, end=' ')
````

### `for` over a set

We cannot take an item of a set by index — a set has no index:

````python
set_example = {'set', 1, 2, 3, 4, 5}
set_example[0]
````

But we can visit every item of a set with a `for` loop without any problem:

````python
set_example = {'set', 1, 2, 3, 4, 5}

for item in set_example:
    print(item, end=' ')
````

### Order: what is guaranteed and what is not

A **sequence** guarantees the order of its items: they come out in exactly the
order in which we put them in. So this result is completely predictable:

````python
list_example = ['fish', 'dog', 'cat']

for item in list_example:
    print(item, end=' ')
````

A **set** gives no such guarantee. The order depends on the hash values of the
items and on the current size of the hash table — everything we discussed in
section 2.8. The items may come out in any order:

````python
set_example = {'fish', 'dog', 'cat'}

for item in set_example:
    print(item, end=' ')
````

Run the cell above, then restart the kernel and run it again. The order can
change, because the hash of a string is different in every run of the
interpreter. Sometimes it will look like the order we wrote — that is luck, not a
rule.

With numbers the effect is easier to see, because the hash of a number never
changes. Here the set clearly does **not** keep the order in which we wrote the
items:

````python
numbers = {10, 1, 7, 3, 5}     # we write them in this order
print(numbers)                 # they are stored in a different one
````

> **Never** write code that depends on the order of a `set`. If you need a
> defined order, sort the items: `for item in sorted(set_example):`.

## 3.6 `range` — counting in a `for` loop

Very often we want to repeat something a fixed number of times, or to work with
numbers. For this Python has `range`:

| Form | Meaning |
|---|---|
| `range(stop)` | `0, 1, ..., stop - 1` |
| `range(start, stop)` | `start, ..., stop - 1` |
| `range(start, stop, step)` | from `start`, in steps of `step`, while we are below `stop` |

The rule is the same as for slices: **`start` is included, `stop` is excluded**.
The step may be negative.

````python
for i in range(5):
    print(i, end=' ')
print()

for i in range(2, 8):
    print(i, end=' ')
print()

for i in range(10, 0, -2):
    print(i, end=' ')
````

`range` does **not** build a list. It is a lazy object that only remembers
`start`, `stop` and `step`, and produces the next number when it is asked. This
is why `range(1000000000)` is created instantly and takes almost no memory. To
see the numbers as a list, we have to ask: `list(range(5))`.

````python
r = range(1000000000)
print(r, type(r))
print(list(range(5)))
````

> **A word about style.** Beginners often write
> `for i in range(len(my_list)):` and then use `my_list[i]`. In Python this is
> almost always the wrong way. If you need the items, loop over the items
> directly: `for item in my_list:`. If you need the index **as well**, use
> `enumerate`.

## 3.7 `enumerate` and `zip`

**`enumerate`** gives us the index *and* the item at the same time. It is the
Python answer to the classic counter loop.

````python
wish_list = ['smart-watch', 'notebook', 'iphone']

for i, item in enumerate(wish_list):
    print(i, item)

print()

for i, item in enumerate(wish_list, start=1):   # numbering can start anywhere
    print(i, item)
````

**`zip`** walks over **several** collections at the same time and gives us the
items in pairs (or in triples, and so on). It stops as soon as the **shortest**
collection ends. We already used `zip` in Part 2 to build a dictionary from two
lists.

````python
names = ['ivan', 'peter', 'maria']
scores = [100, 200, 300]

for name, score in zip(names, scores):
    print(name, '->', score)
````

## 3.8 Loops over dictionaries

A dictionary is iterable, and here we must be careful about **what** it gives us.
By default a `for` loop over a dictionary gives the **keys**:

````python
for key in {'a': 1, 'b': 2}:
    print(key)
````

The same thing, written explicitly with the method `keys()`:

````python
for key in {'a': 1, 'b': 2}.keys():
    print(key)
````

If we need the values, we ask for `values()`:

````python
for value in {'a': 1, 'b': 2}.values():
    print(value)
````

And if we need both at once — which is the usual case — we use `items()`. It
gives a tuple `(key, value)` on every iteration, and we unpack it into two
variables straight away.

````python
for key, value in {'a': 1, 'b': 2}.items():
    print(key, value)
````

And a set, for comparison:

````python
for element in {'a', 'b', 'c'}:
    print(element)
````

> **Important warning.** Do not add or remove keys while you are looping over a
> dictionary or a set. Remember section 2.8: adding an item can make the hash
> table grow and move everything to new slots, and the loop loses its place.
> Python notices this and raises
> `RuntimeError: dictionary changed size during iteration`.
> If you must change the dictionary, loop over a copy of the keys:
> `for key in list(d):`.

## 3.9 Nested loops

The body of a loop is ordinary code, so it may contain another loop. The inner
loop runs **completely** on every iteration of the outer one.

````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, '*', j, '=', i * j, end='   ')
    print()
````

Two things to remember:

* The amount of work **multiplies**. Two nested loops over 1000 items each mean
  1 000 000 iterations.
* `break` and `continue` work only on **their own** loop — the nearest one. A
  `break` in the inner loop does not leave the outer loop.

## 3.10 Comprehensions

We often have to create a **derived** collection — a new collection whose items
are calculated from the items of another one. For example, a list of squares.

With an ordinary loop it looks like this:

````python
input_list = [3, 14, 15, 92, 65, 35, 89, 79, 323, 84626]

output_list = [0] * len(input_list)      # a list of zeros, as long as input_list
for i, item in enumerate(input_list):
    output_list[i] = item ** 2

print(output_list)
````

Python has a much shorter form for exactly this task. It is called a **list
comprehension**:

```python
[expression for variable in iterable]
```

````python
output_list = [item ** 2 for item in input_list]
print(output_list)
````

We can add a **condition**, and then only the items that pass it get into the
result. Here we keep only the even numbers:

```python
[expression for variable in iterable if condition]
```

````python
output_list = [item ** 2 for item in input_list if item % 2 == 0]
print(output_list)
````

Do not confuse the two positions of a condition. The `if` **at the end** filters
items out. A **conditional expression at the beginning** does not filter
anything — it chooses which value to write for every item:

````python
result = [x if x != 7 else 700 for x in range(10)]
print(result)
````

### `set` and `dict` comprehensions

The same syntax works for sets and dictionaries — only the brackets change:

```python
# set comprehension
{value for var in container if condition}

# dict comprehension
{key: value for var in container if condition}
```

````python
result = {x ** 2 for x in range(10)}
print('set comprehension: ', type(result), result)

result = {x: x ** 2 for x in range(10)}
print('dict comprehension:', type(result), result)
````

> **A word about readability.** A comprehension with one loop and one condition
> is beautiful and is considered good Python style. A comprehension with three
> nested loops and a nested conditional expression is a puzzle. Remember the Zen
> of Python: *"Readability counts."* If you cannot read your own line the next
> morning, write an ordinary loop instead.
