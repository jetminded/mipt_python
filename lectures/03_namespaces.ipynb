#  Namespaces, Scopes and Closures


# Namespaces

As you remember, the assignment operator does not create a copy of an object. It only creates a
link between an object in memory (RAM) and a name that is convenient for us. Without a name, it is
very difficult to access the object we need.

For example, the statement

```python
x = 'Hello World!'
```

creates a link between the name `x` and the object `'Hello World!'` in memory.

A **namespace** is a set of names that are defined at a given moment, together with information about
which objects in memory these names refer to.

You can think of a namespace as a large dictionary: the names of variables are the keys, and the objects
in memory are the values. Each key–value pair corresponds to one variable.

A program of any real size creates hundreds or thousands of names, and each name refers to its own
object. Python has to keep track of all these names and make sure that they do not conflict with each other.

Namespaces are not just an important part of Python. Look at the last line of "The Zen of Python":

````python
import this
````

Python has four types of namespaces:

- **built-in**
- **global**
- **enclosing**
- **local**

They have different lifetimes. While a program is running, Python creates new namespaces when they are
needed and deletes them when they are no longer needed. Usually several namespaces exist at the same time.

Let us look at each type of namespace.

## Built-in namespace

The built-in namespace contains the names of all built-in Python objects. These names are available
for the whole time the program is running. We can get a list of all objects in the built-in namespace
with `dir(__builtins__)`:

````python
print(*dir(__builtins__), sep=', ')
````

We will study the `dir` function in more detail in the next lesson.

In this output, we can see many functions that we already know (`print`, `len`, `range`, ...).

The built-in namespace is created when the interpreter starts, and it is deleted only when the
interpreter stops.

## Global namespace

The global namespace contains all names defined at the level of the main program. Python creates it
when the main program starts, and it exists until the interpreter stops.

Strictly speaking, there can be more than one global namespace. The interpreter also creates a separate
global namespace for each module that the program loads with `import`. We will talk about modules in
another lesson.

Let us create several variables and look at the global namespace:

````python
variable_a, variable_b, variable_c = 10, 20, 30
````

````python
print(*dir(), sep=', ')
````

## Local and enclosing namespaces

Namespaces are created not only when the main program starts, but also by parts of the program,
for example by *functions*: a new namespace is created **every time a function is called**.

A **local namespace** is the namespace that contains the names defined inside a function. It is created
when the function is called, and it exists until the function returns a value (including `None`).

Let us check this with an example:

````python
variable = 2000
````

````python
id(variable)
````

````python
def some_func(x):
    i = 1
    variable = 10000
    print(*dir(), sep=', ')
    print("variable =", variable, id(variable))
````

````python
some_func(0)
````

````python
variable
````

````python
id(variable)
````

As you can see, the namespace created by the call of `some_func` contains both the arguments of the
function and the variables defined inside it. The local `variable` is a different object from the global
`variable`: their `id` values are different.

These local variables are not visible in the global namespace:

````python
print(*dir(), sep=', ')
````

It is important to remember that **loops and conditional statements do not create their own namespaces!**

````python
for loop_counter in range(10):
    value_assigned_in_for = 2
````

````python
loop_counter
````

````python
value_assigned_in_for
````

````python
if True:
    value_assigned_in_if = 1
````

````python
value_assigned_in_if
````

If we are not inside a function but in the main program, the local namespace is the same as the global
namespace. Let us check it:

````python
locals() is globals()
````

However, functions do not exist on their own: they must be defined and called somewhere.

The **enclosing namespace** of a function is the namespace of the outer function in which this function
is defined.

Why do we need an extra term for this? Aren't functions always created in the global namespace?

No, this is not true. We can define functions not only in the global namespace (in the main program),
but also inside other functions.

Here is an example of such a **nested function**:

````python
def f():
    print('Start f()')

    def g():
        print('Start g()')
        print('End g()')
        return

    g()

    print('End f()')
    return


f()
````

In this case, the enclosing namespace for the function `g` is the namespace of the function `f`,
not the global namespace.

# Why are nested functions useful?

Nested functions are useful in the following cases:

1. **Encapsulation.** We often need to hide a function from the outer code (the global scope) or to
   protect it from changes.
2. **Internal helper functions.** Sometimes a function needs to repeat the same piece of code several
   times, and this code is not used anywhere else.
3. **Closures** — functions that keep their state between calls.
4. **Decorators** — a way to change the behaviour of a function without changing its code.

# Scope

Because there are several namespaces, several variables with the same name can exist at the same time.
They belong to different namespaces and refer to different objects in memory.

This raises a question: how does Python know which variable we mean when we use its name?

The concept of **scope** answers this question. The scope of a name is the part of the program in which
this name is available. The interpreter determines the scope from the place where the name is defined
and the place where it is used.

## The LEGB rule

When a program uses the name `x`, Python searches for this name in the namespaces in the following order:

- **Local:** if `x` is used inside a function, the interpreter first searches for it in the innermost
  namespace, which is the local namespace of this function.
- **Enclosing:** if `x` is not found in the local namespace and the function is defined inside another
  function, the interpreter searches in the enclosing namespace.
- **Global:** if `x` is still not found, the interpreter searches in the global namespace.
- **Built-in:** finally, the interpreter searches for `x` in the built-in namespace.

This search order is called the **LEGB rule**.

**What happens if the name is not found even in the built-in namespace?** The interpreter raises a
`NameError`.

Let us look at this rule with some examples.

## Examples of the LEGB rule

### Example 1. One definition

The variable is defined in only one place — in the global namespace:

````python
x = 'global'

def f():

    def g():
        print(x)

    g()


f()
````

The `print()` function can use only one `x`, so it prints the value `global`, which is defined in the
global namespace.

### Example 2. Two definitions

The variable `x` is defined in two places: outside the function `f`, and inside `f` but outside `g`:

````python
x = 'global'

def f():
    x = 'enclosing'

    def g():
        print(x)

    g()


f()
````

````python
print(x)
````

According to the LEGB rule, the interpreter first looks in the enclosing namespace and only after that in
the global namespace. That is why we get the string `enclosing` and not `global`.
The global `x` does not change.

### Example 3. Three definitions

Now we define the variable `x` in three places: outside `f`, inside `f` but outside `g`, and inside `g`:

````python
x = 'global'

def f():
    x = 'enclosing'

    def g():
        x = 'local'
        print(x)

    g()


f()
````

There is a definition of `x` inside `g`, so `x` is a local variable at this place. The LEGB rule looks in
the local namespace first, so `print` outputs the string `local`.

Note: the local assignment must come **before** `print(x)`. If we swap these two lines, we get an error.
We will explain the reason in the section "Python scopes are static".

### Example 4. No definition

Finally, let us see what happens if the variable `x` is not defined anywhere. We defined `x` in the
global namespace earlier, so first we delete it:

````python
del x
````

````python
def f():

    def g():
        print(x)

    g()


f()
````

`x` was not found in any namespace, so we got a `NameError`.

### Example 5. A name can hide a built-in function

The LEGB rule works for all names, including function names. In the example below, the inner function
`print` hides the built-in `print` inside `f`. Inside our `print`, the statement `global print` makes
Python look for the name in the global namespace. There is no global `print`, so Python finally finds the
built-in function.

````python
def f():
    x = 0

    def print(x):
        global print
        print(x, x)

    def g():
        print(x)

    g()
````

````python
f()
````

````python
print(0)
````

### Example 6. Changing an object vs. creating a new name

Calling a method of a mutable object (for example, `append`) is not an assignment. Python finds the global
list `l` with the LEGB rule and changes this object:

````python
l = [0, 1]
def f():
    def g():
        l.append(0)
        print(l)

    g()
````

````python
f()
````

````python
l
````

An assignment, however, creates a new **local** name. The global list does not change:

````python
l = [0, 1]
def f():
    def g():
        l = [0]
        print(l)

    g()
````

````python
f()
````

````python
l
````

# Namespace dictionaries

Earlier we said that a namespace can be seen as a dictionary in which the key is the name of a variable
and the value is the object.

For the global and local namespaces this is literally true: Python really stores these namespaces as
dictionaries.

## The `globals` function

The built-in function `globals()` returns a reference to the dictionary that contains the current global
namespace. We can use it to access the elements of the global namespace.

Let us look at an example:

````python
type(globals())
````

````python
globals()
````

As you can see, the interpreter has already put some variables into the global namespace. These variables
may be slightly different depending on the Python version, the operating system and other software.

Now let us see how the dictionary changes when we define a new variable:

````python
x = 'foo'
````

````python
globals()
````

After the assignment `x = 'foo'`, a new element appeared in the dictionary of the global namespace.

Usually we access an object through the name of a variable. However, we can also read, change and add
variables through the dictionary of the global namespace:

````python
x
````

````python
globals()['x']
````

````python
x is globals()['x']
````

Let us add a new variable:

````python
globals()['y'] = 100
y = 100
````

````python
'y' in globals()
````

````python
y
````

````python
globals()['y'] = 3.14159
````

````python
y
````

Since `globals()` is an ordinary dictionary, we can even use a key that is not a valid variable name,
for example a tuple. Such a key can be read only through the dictionary, not as a variable:

````python
globals()[(1,2,3)] = 3.14159
````

````python
globals()[(1,2,3)]
````

````python
(1,2,3)
````

## The `locals` function

There is a similar function for the local namespace: `locals()`.

````python
def f(x, y):
    s = 'foo'
    print(locals())
````

````python
f(10, 0.5)
````

When `locals()` is called inside the function `f()`, it returns a dictionary that represents the local
namespace of this function. Note that, in addition to the locally defined variable `s`, the dictionary
contains the arguments `x` and `y`, because they are also local names of `f()`.

If we call `locals()` outside a function, in the main program, we get the same dictionary as `globals()`:

````python
globals() is locals()
````

# The keywords `global` and `nonlocal`

## The keyword `global`

Sometimes we need to change a variable in the global scope from inside a function. In Python, we can do
this with the keyword `global`:

````python
x = 20
y = 10

def f():
    global x, y
    x = 40
    print(x, y)
````

````python
x
````

````python
f()
````

````python
x
````

The statement `global x` means that, while the function `f()` is running, every use of the name `x`
refers to `x` in the global namespace.

So the assignment `x = 40` does not create a new local variable. It changes the variable `x` in the
global scope.

We can get the same result with the function `globals()`:

````python
x = 20

def f():
    globals()['x'] = 40
    print(x)
````

````python
x
````

````python
f()
````

````python
x
````

There is no fundamental difference between these two approaches, but the keyword `global` makes the
code cleaner and easier to read.

If the name does not exist in the global namespace yet, the keyword `global` together with an assignment
creates it:

````python
'z' in globals()
````

````python
def g():
    global z
    z = 20
````

````python
g()
z
````

## The keyword `nonlocal`

Nested functions have a similar situation. The keyword `global` allows a function to access and change
an object in the global scope. But what if a nested function needs to change a variable in the
enclosing scope? Let us look at an example:

````python
def f():
    x = 20

    def g():
        x = 40

    g()
    print(x)
````

````python
f()
````

Here the assignment `x = 40` creates a new local variable inside `g()`. The variable `x` in the enclosing
scope does not change: the function `g()` cannot change a value in the global or the enclosing scope
directly.

The keyword `global` does not help here either:

````python
del x
````

````python
def f():
    x = 20

    def g():
        global x
        x = 40

    g()
    print(x)
````

````python
f()
````

````python
x
````

The function `f()` still prints `20`, and `global x` has created a **new global** variable `x = 40`.
This is not what we wanted.

For such situations we use the keyword `nonlocal`. The interpreter then looks for the variable in the
nearest enclosing namespace:

````python
def f():
    x = 20

    def g():
        nonlocal x
        x = 40

    g()
    print(x)
````

````python
f()
````

Note: `nonlocal` works only with enclosing scopes of functions. If the name is not defined in any
enclosing function, Python raises a `SyntaxError` already when it reads the code.

# Python scopes are static

Python scopes are **static**. This means that Python analyses the code of a function and decides which
names are local **before** the code is executed. Let us look at an example:

````python
x = 1

def f():
    print(x)

    def g():
        print(x)

    g()
    x = 2

f()
````

There is an assignment to `x` in the function `f`, so Python decides that `x` is a local variable of
`f` everywhere in this function. When we call `print(x)` earlier in the same function, the local variable
does not have a value yet, and we get an `UnboundLocalError`.

We get a similar error if we comment out the first `print(x)`:

````python
x = 1

def f():
    # print(x)
    x = 2
    def g():
        print(x)

    g()
    x = 2

f()
````

Python knows that the local variable `x` will be created in the function `f`. So when the nested
function `g` tries to print `x`, Python looks for `x` in the enclosing scope of `g` (the scope of `f`).
In this scope the variable exists but does not have a value yet: it will get a value later.

# Closures

> In computer programming languages, a closure is a function together with a referencing environment of
> that function. A closure function is any function that uses a variable that is defined in an
> environment (or scope) that is external to that function, and is accessible to the function when it is
> invoked from a scope in which that free variable is not defined.

A **closure** is a function whose body uses variables that are defined outside this function, in the
surrounding code, and that are not its parameters. Such variables are called **free variables**.

Closures exist in Python because of three things: the LEGB rule, the fact that functions are objects,
and the fact that scopes in Python are static.

Let us see why it is important to understand closures. We want to create a list of functions, and each
function multiplies its argument by its own factor:

````python
multipliers = []

for m in range(5):
    multipliers.append(lambda x: x * m)
````

````python
multipliers
````

````python
print('m =', m)
````

````python
[multipliers[i](5) for i in range(5)]
````

````python
multipliers[0]
````

````python
multipliers[0](5)
````

Why do we get `[20, 20, 20, 20, 20]` and not `[0, 5, 10, 15, 20]`?

Our lambda functions do not define the variable `m`, so the interpreter looks for it (and finds it) in the
global scope. When we call the functions, they use the global variable `m`, and its value may change
between the moment the function is created and the moment it is called. As a result, all our lambda
functions refer to the same variable `m`, which is a global variable. After the loop, `m` is equal to `4`.

So, if we change the variable `m`, the behaviour of our functions also changes:

````python
m = 0

print([multipliers[i](5) for i in range(5)])
````

In fact, this is equivalent to the following example:

````python
def fn1(x):
    print(m)
    return m * x

def fn2(x):
    print(m)
    return m * x

def fn3(x):
    print(m)
    return m * x

multipliers = [fn1, fn2, fn3]

m = 5

print([multipliers[i](5) for i in range(3)])
````

````python
multipliers[0]
````

Let us look at an even simpler example:

````python
def function(a, b):
    return NAME, a, b
````

````python
function(1, 2)
````

````python
NAME = 'Alice'
function(1, 2)
````

````python
NAME = 'Bob'
function(1, 2)
````

The function does not store the value of `NAME`. It looks the name up every time it is called.

If we want to use variables that are defined outside a function but do not want to keep them in the
global scope, we can use the following construction:

````python
def foo():

    x = 3

    def bar():
        print(x)
    bar()
    x = 5
    return bar
````

````python
bar_global = foo()
bar_global()
````

````python
x = 9
bar_global()
print('x =', x)
````

Now the function `bar` refers to the variable `x` from the enclosing scope. That is why its behaviour
does not change when we change the global variable `x`.

Note that `bar_global()` prints `5`, not `3`. A closure remembers the **variable**, not its value at the
moment when the inner function was defined. The value is taken when the function is called, and at that
moment `x` in `foo` is already equal to `5`.

````python
from datetime import datetime
from time import sleep

def make_remembering_function():
    time_of_creation = datetime.now()
    def remembering_function():
        print(f"I was created at {time_of_creation}")
    return remembering_function

f1 = make_remembering_function()
sleep(1)
f2 = make_remembering_function()

for f in (f1, f2):
    f()
````

First, let's note that inside the body of this function, another function called `remembering_function` is defined, and it is also returned as the result. Each time the `make_remembering_function` function is called, the interpreter encounters the definition of `remembering_function` again, creates a function object, and binds it to the name `remembering_function` in the local namespace of the `make_remembering_function` function (each call to the same function has its own unique local namespace).

Now let's note that the created function `remembering_function` accesses the variable `time_of_creation`, which is not defined inside this function.

Such variables are considered free variables, and you can even access the list of them directly.



````python
f1.__code__.co_freevars
````

The binding of free variables to objects happens at the time the function is called. The lookup is performed in the following order of priority:

- **nonlocals** — the local namespace of the enclosing function (in this example, the local namespace of the `make_remembering_function` function);
- **globals** — the global namespace of the module in which this function is defined;
- **builtins** — the namespace of built-in names.

You can also perform the resolution process artificially using the `inspect` module, which is designed for inspecting Python objects. The `inspect.getclosurevars` method resolves all names that are used in a function but not declared within the function itself, matching them to objects.

````python
import inspect

print(inspect.getclosurevars(f1))
print(inspect.getclosurevars(f2))
````

A closure can read a free variable, but an assignment inside the inner function makes the name local.
In the next example, `x += 1` is an assignment, so we get an error:

````python
def outer():
    x = 1
    def inner():
        # x = 0
        print(f'x in outer function (before modifying): {x}')
        x += 1
        print(f'x in outer function (after modifying): {x}')
    return inner
````

````python
f = outer()
f()
````

To change the free variable, we use `nonlocal`:

````python
def outer():
    x = 1
    def inner():
        nonlocal x
        print(f'x in outer function (before modifying): {x}')
        x += 1
        print(f'x in outer function (after modifying): {x}')
    return inner
````

````python
f = outer()
f1 = outer()
f()
````

````python
f()
````

The value of `x` is kept between the calls: this is how a closure stores its **state**.

A classic example of a closure is a function factory — a function that creates and returns other
functions:

````python
def make_adder(x):
    def adder(y):
        return x + y
    return adder
````

````python
add_two = make_adder(2)

print(add_two(5))
print(add_two(7))
````

````python
add_three = make_adder(3)

add_three(5)
````

````python
add_two(5)
````

Every call of `make_adder` creates a new namespace, so `add_two` and `add_three` do not affect each other.

An outer function can return several closures at once:

````python
def make_adder(x):
    def adder(y):
        return x + y
    def adder1(z):
        return x + z * 2
    return adder, adder1
````

````python
adder_global, adder1_global = make_adder(42)
````

````python
adder_global(1)
````

````python
adder1_global(1)
````

Several functions can share the same free variable:

````python
def cell1(value=0):
    def Get():  # Functions are usually named in lowercase, like variables. Here we break this rule to distinguish inner functions from outer ones
        return value

    def Set(new_value):
        nonlocal value
        value = new_value

    return Get, Set
````

````python
Get, Set = cell1(10)
print(Get())
````

````python
Set(20)
print(Get())
````

The variable `value` is not available in the global scope:

````python
value
````

Let us see what is stored inside a closure:

````python
print(Get.__closure__, type(Get.__closure__))
print(Get.__closure__[0].cell_contents)
````

**`__closure__`** is a tuple of the free variables of a function (or `None` if the function has no free
variables). <br>
Each variable is stored as an object of the **cell** class, which has one attribute: **`cell_contents`**.

The names of the free variables are stored in `__code__.co_freevars`:

````python
Get.__code__.co_freevars
````

`Get` and `Set` share the same cell object, so a change made by `Set` is visible to `Get`:

````python
print(Get.__closure__ == Set.__closure__)
print(Get.__closure__[0] is Set.__closure__[0])
````
