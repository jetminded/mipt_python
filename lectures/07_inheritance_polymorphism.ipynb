# Object-Oriented Programming in Python

# What is OOP?

**A programming paradigm** is a set of ideas and rules that defines a style of writing programs. In other words, it is a general approach to programming.

**The main programming paradigms:**
- **imperative programming** — the program is a sequence of commands that run one after another;
- **procedural programming** — we put repeated code into procedures and functions, so we do not write the same code again;
- **declarative programming** — the program describes *what* result we need, not *how* to get it (SQL is a good example);
- **functional programming** — the program is a set of functions; the output of one function becomes the input of another;
- **structured programming** — an important step in the history of programming; it describes a program with a few basic structures: sequence, branching (`if/else`), loops (`while`) and subroutines (functions);
- **object-oriented programming (OOP)** — the topic of this seminar.

**Object-oriented programming** is a programming paradigm in which a program is a set of objects that interact with each other. Each object has its own properties (data) and behavior (actions). Each object is an instance of a class, and classes can form an inheritance hierarchy.

**The main terms of OOP:**
- class;
- object;
- abstraction;
- encapsulation;
- inheritance;
- polymorphism.

## Objects and classes

The main idea of OOP is to divide a program into separate entities — **objects**.

**Examples of objects in real life:** a table, a cup, a person, a book, a building.
**Examples of objects in programs:** a user, a button, a message.

Objects have properties, for example, a color, some content or a user name. These variables (the characteristics of a particular object) are called **attributes**.

The actions of an object, for example, "send a message", "hire an employee" or "click a button", are called **methods**. A method is a function that is defined inside a class. It belongs to an object and works with the data of this object or with other parts of the program.

We do not want to write separate methods for every object, because we want to reuse our code. For this reason, we use classes. A **class** is a template for objects. It describes the attributes and methods that the objects will have. Every object is an **instance** of some class.

**Example.** We are developing a system for managing employees. We create the class `Employee` with the following attributes: full name, position, department, manager and salary. The class also has the following methods: "move to another department", "raise the salary", "change the position" and "dismiss". A particular employee, Vasya Petrov, is an object, that is, an instance of the class `Employee`.

## Abstraction

**Abstraction** means that we select the most important attributes and methods of an object and leave out the unimportant ones. We describe an object with the smallest possible set of attributes and methods, but this set must still be enough to solve the task.

**Example.** An object of the class `Programmer` probably does not need the attributes "can cook" or "favorite color", because they are not connected with programming. However, "main programming language" and "work skills" are important attributes: we cannot describe a programmer without them.

The set of attributes and methods that is available from outside works as the **interface** of the object. Other parts of the program use this interface to work with the object. They do not need to know how the object works inside.

## Encapsulation

The word **encapsulation** has two connected meanings.

1. **Data and methods are kept together.** Every object is an independent structure: it contains all the data and methods that it needs. If an object needs a variable, this variable is stored inside the object, not in the outside code. As a result, objects are more flexible: if someone changes the outside code, the logic of the object stays the same.
2. **The internal details are hidden.** Other objects do not need to know *how* an object does its job; it is only important that it does the job correctly. From outside, we can see only the values of public attributes and the results of methods. The internal data of an object is closed to other objects.

Encapsulation protects an object from incorrect use. For example, a bank account should not allow anyone to set a negative balance directly.

## Inheritance

**Inheritance** allows us to create a new class on the basis of an existing class. The new class gets all the attributes and methods of the existing class, and it can add new ones or change some of them. The more general class is called the **parent** (base class, superclass), and the more specific class is called the **child** (derived class, subclass).

**Example.** The class `Employee` can have the child classes `Programmer`, `Analyst` and `Tester`. Each of them has its own methods, for example, "write code", "analyze" and "test". In addition, all the methods and attributes of the class `Employee` are available in all the child classes. Our Vasya Petrov is an instance of both the class `Programmer` and the class `Employee`.

One parent class can have several child classes. In Python, one child class can also have several parent classes (this is called **multiple inheritance**).

Inheritance helps us build a clear hierarchy of classes "from general to specific". This makes the code easier to understand and to extend. We do not need to write the same attributes and methods again in different classes: we define them once in the parent class, and all the child classes get them automatically.

## Polymorphism

The word **polymorphism** comes from Greek and means "many forms". In OOP, it means that methods with the same name can work differently in different classes. For example, the class `Employee` has the method "work". For a programmer, this method means "write code", and for a tester it means "test the program". However, in both cases it is work.

Polymorphism allows us to write general code: we call the method `work()` for every employee, and each object does it in its own way.

For this reason, common rules and agreements between developers are important. If a method is called `delete`, it must delete something. The way it deletes depends on the object, but the method must perform deletion. In addition, if the team has agreed that the method for deleting is called `delete`, we should not call it `remove` or something else in another class, because this makes the code confusing.

# OOP in Python

## Types of methods

As we already know, functions inside a class are called **methods**. There are three types of methods:
- instance methods (normal methods);
- static methods;
- class methods.

Let us see the difference between them.

### 1. Instance methods (normal methods)

These methods can be used only after an instance of the class is created: we call such a method for an instance. Therefore, the first parameter of such a method is `self`. As we discussed above, it receives a reference to the object for which the method was called. Here is an example:

````python
class Phone:

    def __init__(self, color, model):
        self.color = color
        self.model = model

    # A normal method
    # The first parameter of the method is self
    def check_sim(self, mobile_operator):
        if self.model == 'I785' and mobile_operator == 'MTS':
            print('Your mobile operator is MTS')
````

````python
# Create an instance of the class
my_phone = Phone('red', 'I785')

# Call the method check_sim() on the object my_phone
my_phone.check_sim('MTS')
````

Usually this is the largest and the most frequently used group of methods (compared with static methods and class methods).

### 2. Static methods

Static methods are normal functions that we place inside a class for convenience, so they belong to the namespace of this class. Usually they contain some helper code.

The important point is that we can call these methods with the name of the class, and we do not need to create an object. That is why these methods do not have the parameter `self`: they do not need any information about the objects of the class.

Do you remember the method `method_without_self` from the class `MyLittleClass4`? It worked only when we called it for the class. The correct way to write such methods is to use the special decorator `@staticmethod`. Then the method works both for the class and for an instance:

````python
class Phone:

    # A static method for reference
    # It returns a hash by the model number
    # There is no self in the method
    @staticmethod
    def model_hash(model):
        if model == 'I785':
            return 34565
        elif model == 'K498':
            return 45567
        else:
            return None

    # A normal method
    def check_sim(self, mobile_operator):
        pass
````

````python
# Call the static method model_hash using only the class name
# We do not need to create a Phone object
print(Phone.model_hash('I785'))

# We can call it on an instance too
print(Phone().model_hash('K498'))
````

### 3. Class methods

Class methods are something between normal methods (they are bound to an object) and static methods (they are bound only to the namespace of the class). As the name suggests, these methods are closely connected with the class in which they are defined.

These methods can change the state of the class itself, and this affects ALL instances of the class. However, they cannot change one particular object of the class (instance methods do this).

To create a class method, we use the decorator `@classmethod`. The first parameter of such a method is `cls`. Unlike `self`, `cls` is a reference to the class itself, not to an object. Let us look at an example:

````python
class Phone:

    def __init__(self, color, model, os):
        self.color = color
        self.model = model
        self.os = os

    # A class method
    # It takes 1) a reference to the class Phone and 2) a color as parameters
    # It creates a special Phone object (the special thing is that it is a toy phone)
    # To do this, it calls the initializer of the class Phone
    # and passes the color and the model of the toy phone to it
    @classmethod
    def toy_phone(cls, color):
        toy_phone = cls(color, 'ToyPhone', None)
        return toy_phone

    # A static method
    @staticmethod
    def model_hash(model):
        pass

    # A normal method
    def check_sim(self, mobile_operator):
        pass
````

````python
# Create a toy phone object
# Call the class method toy_phone using the class name and a dot
my_toy_phone = Phone.toy_phone('Red')
print(my_toy_phone.color, my_toy_phone.model, my_toy_phone.os)
````

In the following example, a class method changes the state of the class, so ALL instances see the change:

````python
class Phone:
    default_color = 'Grey'

    def __init__(self, model):
        self.model = model

    # A class method: it changes the class attribute for all instances
    @classmethod
    def set_default_color(cls, color):
        cls.default_color = color


phone1 = Phone('I495')
phone2 = Phone('C385')
Phone.set_default_color('Black')
print(phone1.default_color, phone2.default_color)
````

As the examples show, class methods are often used when:
- we need to create a special object of the current class;
- we use the factory pattern, that is, we create objects of different child classes inside the method.

**Summary:**

| Type of method | Decorator | First parameter | How to call it |
|---|---|---|---|
| instance method | — | `self` (the object) | `obj.method()` |
| static method | `@staticmethod` | — | `Class.method()` or `obj.method()` |
| class method | `@classmethod` | `cls` (the class) | `Class.method()` or `obj.method()` |

## Encapsulation in Python: `_` and `__`

In many programming languages (for example, C++ and Java), we can control access to the attributes of a class from the outside code and from other classes. Usually there are three access levels:
- **public** — the attributes are available inside the class and from the outside code;
- **protected** — the attributes are available inside the class and in its child classes;
- **private** — the attributes are available only inside the class.

In these languages, special keywords (`public`, `protected`, `private`) control access, and the compiler does not allow access that breaks these rules.

Does Python have such access control? Can we stop the outside code from reading and changing the attributes of an object? Python has no special keywords for this. Instead, it uses the number of **underscores** at the beginning of a name:

````python
class VeryPrivateDataHolder:
    not_secret = 0     # public
    _secret = 1        # protected
    __very_secret = 2  # private
````

````python
obj = VeryPrivateDataHolder()
````

````python
obj.not_secret
````

````python
obj._secret
````

````python
obj.__very_secret
````

At first sight, Python does have private attributes. However, the attribute still exists; it only has a different name:

````python
obj._VeryPrivateDataHolder__very_secret
````

So we can still access "private" attributes if we really want to. However, this is not recommended, especially for classes that other people have written.

````python
obj._VeryPrivateDataHolder__very_secret = 'new secret'
obj._VeryPrivateDataHolder__very_secret
````

````python
VeryPrivateDataHolder.__very_secret
````

### Naming conventions

Python has no strict access control. A well-known principle of the Python community says: *"We are all consenting adults here"*, that is, people who can make their own decisions. It means that the programmer is responsible for the correct use of a class. The name of an attribute only shows the intention of its author:

| Name | Access level | Meaning |
|---|---|---|
| `name` | public | a part of the interface of the class; anyone can use it |
| `_name` | protected | an internal attribute; do not use it outside the class and its child classes |
| `__name` | private | an internal attribute of this class only; Python changes its name (**name mangling**) |
| `__name__` | special | a special attribute or method of Python (for example, `__init__`); Python does not change the name |

Let us look at each case in more detail.

### One underscore: `_name` (protected)

One underscore at the beginning of a name is **only an agreement** between programmers. It means: "This is an internal part of the class. Do not use it from outside, because it can change in the next version of the code." Python itself does not stop us:

````python
class Thermometer:
    def __init__(self, celsius):
        self._celsius = celsius   # protected: for internal use

    def show(self):
        print(f'Temperature: {self._celsius} °C')


t = Thermometer(21)
t.show()
print(t._celsius)  # this works, but we should not do this
````

Code editors (for example, PyCharm or VS Code) often show a warning when we use a protected attribute outside its class. In addition, the statement `from module import *` does not import the names that start with an underscore.

### Two underscores: `__name` (private) and name mangling

If a name inside a class starts with two underscores (and does not end with two underscores), Python changes this name automatically: it adds an underscore and the name of the class at the beginning. For example, the attribute `__very_secret` of the class `VeryPrivateDataHolder` is stored as `_VeryPrivateDataHolder__very_secret`. This process is called **name mangling**.

Inside the class, we use the short name, and Python changes it for us. We can see the real names of the attributes in the dictionary `__dict__` of the object:

````python
class User:
    def __init__(self, login, password):
        self.login = login            # public
        self.__password = password    # private

    def check_password(self, password):
        # inside the class, we use the short name
        return self.__password == password


user = User('anna', 'qwerty')
print(user.__dict__)
print(user.check_password('12345'))
print(user.check_password('qwerty'))
````

Name mangling happens **only inside the body of a class**. If we write `user.__password = ...` outside the class, Python creates a new normal attribute with the name `__password`. The private attribute does not change:

````python
user.__password = '00000'            # outside the class: this is a new attribute
print(user.__dict__)
print(user.check_password('00000'))  # the real password has not changed
````

Names that **start and end** with two underscores (for example, `__init__`, `__dict__`, `__len__`) are not changed. Python uses such names for its special attributes and methods, so do not create new names of this kind.

**Why does Python need name mangling?** Its main purpose is not to hide data from other programmers. It protects the attributes of a class from **name conflicts** with its child classes. We will see an example in the section about inheritance.

### Getters, setters and `property`

Encapsulation is not only about hiding data. It also helps us **control** how the data changes. For example, the balance of a bank account cannot be negative. If the attribute is public, anyone can save a wrong value:

````python
class Account:
    def __init__(self, balance):
        self.balance = balance


acc = Account(100)
acc.balance = -500   # nothing stops us
print(acc.balance)
````

The classic solution is to make the attribute private and to write two methods: a **getter** (it returns the value) and a **setter** (it checks the new value and saves it):

````python
class Account:
    def __init__(self, balance):
        self.set_balance(balance)

    def get_balance(self):         # getter
        return self.__balance

    def set_balance(self, value):  # setter
        if value < 0:
            raise ValueError('The balance cannot be negative')
        self.__balance = value


acc = Account(100)
acc.set_balance(250)
print(acc.get_balance())
acc.set_balance(-500)
````

This code works, but it is not convenient: instead of `acc.balance` we must write `acc.get_balance()`. Python has a better solution, the decorator **`@property`**. It turns a method into an attribute: we read and change it with a dot, but each time Python calls our methods:
- the method with `@property` is the **getter**; it runs when we read the attribute;
- the method with `@<name>.setter` is the **setter**; it runs when we assign a value to the attribute;
- the method with `@<name>.deleter` runs when we delete the attribute with `del` (it is rarely used).

````python
class Account:
    def __init__(self, balance):
        self.balance = balance         # this line calls the setter

    @property
    def balance(self):                 # getter
        return self.__balance

    @balance.setter
    def balance(self, value):          # setter
        if value < 0:
            raise ValueError('The balance cannot be negative')
        self.__balance = value


acc = Account(100)
acc.balance = 250        # the setter is called
print(acc.balance)       # the getter is called
acc.balance = -500       # the setter raises an error
````

If a property has no setter, it is **read-only**. A property can also be **computed** from other attributes, so we do not need to store its value:

````python
import math


class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value <= 0:
            raise ValueError('The radius must be positive')
        self._radius = value

    @property
    def area(self):          # a computed property without a setter
        return math.pi * self._radius ** 2


c = Circle(2)
print(c.area)
c.radius = 3             # we change the radius, and the area changes too
print(c.area)
c.area = 100             # read-only: there is no setter
````

Note that the class `Circle` stores the value of the property in the protected attribute `_radius`. This is a common practice: the property `radius` is the public interface, and `_radius` is the internal storage. Also note that `__init__` uses the property (`self.radius = radius`), so the check works when the object is created, too.

### `__slots__`: a fixed set of attributes

By default, we can add any new attribute to an object (see the section "Dynamic attributes"). Sometimes this is dangerous: a mistake in the name of an attribute creates a new attribute, and Python shows no error. The class attribute `__slots__` lists all the attributes that an object can have. Python does not allow us to create any other attributes:

````python
class Point:
    __slots__ = ('x', 'y')

    def __init__(self, x, y):
        self.x = x
        self.y = y


p = Point(1, 2)
p.x = 10        # correct: 'x' is in __slots__
p.z = 3         # error: 'z' is not in __slots__
````

Objects with `__slots__` do not have the dictionary `__dict__`, so they also use less memory. This is useful when a program creates millions of small objects.

**Summary:**
- Python has no keywords like `private` or `protected`. Access levels are based on naming conventions.
- `name` is public; `_name` is protected (it is only an agreement); `__name` is private (Python changes the name to `_ClassName__name`).
- Names like `__init__` are special names; Python does not change them.
- `@property` allows us to control the reading and writing of an attribute and to keep the simple syntax `obj.attr`.
- `__slots__` fixes the set of attributes that an object can have.

## Inheritance

### The syntax of inheritance

To create a child class, we write the name of the parent class in brackets after the name of the new class:

```python
class Child(Parent):
    ...
```

The parent class is also called the **base class** or the **superclass**. The child class is also called the **derived class** or the **subclass**.

As we already know, every class in Python inherits from `object`, even if we do not write it. The attribute `__bases__` shows the direct parents of a class, and the function `issubclass(A, B)` checks whether the class `A` is a child of the class `B`:

````python
class MyClass:
    pass

print(MyClass.__bases__)
print(issubclass(MyClass, object))
````

Let us see how inheritance works in Python. We will use the following classes:

````python
class Animal:  # implicitly inherits from the class object
    some_value = "animal"

    def __init__(self):
        print("i am an animal")

    def speak(self):
        raise NotImplementedError('i don\'t know how to speak')  # an error that shows that the method is not implemented yet
````

The statement `raise` raises the given exception. In this way, the programmer can raise a chosen exception on purpose.

The statement `raise` without an exception raises the last active exception again (in the current scope).

````python
class Cat(Animal):
    some_value = "cat"

    def __init__(self):
        super().__init__()
        print("i am a cat")

    def speak(self):
        print('meoooow')
````

````python
class Dog(Animal):
    some_value = "dog"

    def __init__(self):
        super().__init__()
        print("i am a dog")
````

````python
class Hedgehog(Animal):
    def __init__(self):
        super().__init__()
        print("i am a hedgehog")
````

````python
animal = Animal()
print(animal.some_value)
animal.speak()
````

````python
cat = Cat()
````

````python
cat.some_value # overridden
````

````python
cat.speak()
````

````python
dog = Dog()
````

````python
dog.some_value # overridden
````

````python
dog.speak()
````

````python
hedgehog = Hedgehog()
````

````python
hedgehog.some_value
````

````python
hedgehog.speak()
````

### How Python finds attributes

The classes `Dog` and `Hedgehog` do not have the method `speak`, so Python takes it from the parent class `Animal`. The class `Hedgehog` also does not have the attribute `some_value`, so `hedgehog.some_value` returns `"animal"`.

When we write `obj.x`, Python searches for the name `x` in the following order:
1. in the object itself (in its dictionary `obj.__dict__`);
2. in the class of the object;
3. in the parent class, then in the parent of the parent, and so on, up to `object`.

Python stops at the first place where it finds the name. If the name is not found anywhere, Python raises `AttributeError`.

### `isinstance` and `issubclass`

An object of a child class is also an object of the parent class: a cat **is an** animal. This is the main rule of inheritance (the "is-a" relationship). The function `isinstance(obj, Class)` takes inheritance into account, but the comparison `type(obj) == Class` does not:

````python
print(isinstance(cat, Cat), isinstance(cat, Animal), isinstance(cat, object))
print(type(cat) == Cat, type(cat) == Animal)
print(issubclass(Cat, Animal), issubclass(Animal, Cat))
````

Therefore, to check the type of an object, we should use `isinstance`. Then our code works correctly with the parent class and with all its child classes. The second argument can also be a tuple of classes: `isinstance(x, (int, float))` returns `True` if `x` is an `int` or a `float`.

### Overriding and extending methods

A child class can do three things with a method of the parent class:
1. **inherit** it — use it without changes (as `Dog` uses `speak` from `Animal`);
2. **override** it — replace it with its own version (as `Cat` replaces `speak`);
3. **extend** it — call the version of the parent class and add new actions (as all the child classes do with `__init__`).

To call a method of the parent class, we use the function **`super()`**. It returns a special object, and through this object we call the methods of the parent class. Note that we do not pass `self` to the method: we write `super().__init__()`, not `super().__init__(self)`.

Let us return to our example with employees:

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def work(self):
        return f'{self.name} is working'

    def info(self):
        return f'{self.name}, salary: {self.salary}'


class Programmer(Employee):
    def __init__(self, name, salary, language):
        super().__init__(name, salary)   # extend: the parent sets name and salary
        self.language = language         # and we add a new attribute

    def work(self):                      # override: a new version of the method
        return f'{self.name} is writing code in {self.language}'

    def info(self):                      # extend: we use the result of the parent method
        return super().info() + f', language: {self.language}'


vasya = Programmer('Vasya Petrov', 3000, 'Python')
print(vasya.work())
print(vasya.info())
print(isinstance(vasya, Employee))
````

We can also call a method of the parent class directly by the name of the class: `Employee.__init__(self, name, salary)`. In this case, we must pass `self` ourselves. This works in simple cases, but `super()` is better: we do not repeat the name of the parent class, and `super()` works correctly with multiple inheritance (see below).

### Inheritance and encapsulation

Now we can see how `_` and `__` work with inheritance. A protected attribute (`_name`) is available in child classes in the usual way. A private attribute (`__name`) is not available by its short name, because Python has changed its name to `_Parent__name`:

````python
class Parent:
    def __init__(self):
        self._protected = 'protected'
        self.__private = 'private'      # it is stored as _Parent__private


class Child(Parent):
    def show(self):
        print(self._protected)          # this works
        print(self.__private)           # Python looks for _Child__private


child = Child()
child.show()
````

This is exactly the purpose of name mangling. A child class can have its own private attribute with the same name, and it does not break the parent class:

````python
class Engine:
    def __init__(self):
        self.__power = 100              # stored as _Engine__power

    def engine_power(self):
        return self.__power


class TurboEngine(Engine):
    def __init__(self):
        super().__init__()
        self.__power = 150              # stored as _TurboEngine__power: a different attribute

    def turbo_power(self):
        return self.__power


engine = TurboEngine()
print(engine.engine_power())   # the parent method uses the attribute of the parent
print(engine.turbo_power())    # the child method uses the attribute of the child
print(engine.__dict__)
````

If the attribute were protected (`_power`), the child class would overwrite the value of the parent, and the method `engine_power()` would return 150.

**Rule:** use `_name` for attributes that child classes may need. Use `__name` only when you want to protect an attribute from accidental conflicts with child classes.

### Protected and private methods

The same rules work for methods. A protected method (`_method`) is a helper method for the class and its child classes. A private method (`__method`) is a helper method for this class only.

There is an important difference: a child class **can override** a protected method, but it **cannot override** a private method. The parent class always calls its own private method, because Python has changed its name to `_Parent__method`:

````python
class Report:
    def build(self):
        return self._header() + ' | ' + self.__footer()

    def _header(self):          # protected: child classes may change it
        return 'Report'

    def __footer(self):         # private: only for the class Report
        return 'end of report'


class SalesReport(Report):
    def _header(self):          # overrides the protected method
        return 'Sales report'

    def __footer(self):         # does NOT override: this is _SalesReport__footer
        return 'end of sales report'


print(Report().build())
print(SalesReport().build())
````

The method `build()` calls `self._header()`, and Python finds the new version in `SalesReport`. However, `self.__footer()` inside the class `Report` means `self._Report__footer()`, so the method of the child class (`_SalesReport__footer`) is never called.

**Where can we use an attribute or a method?**

| Name | Inside the class | In a child class | Outside the class |
|---|---|---|---|
| `name` | yes | yes | yes |
| `_name` | yes | yes | possible, but we should not do it |
| `__name` | yes | no (only as `_Parent__name`) | no (only as `_ClassName__name`) |

### Multiple inheritance

A class can have several parents. We list them in brackets and separate them with commas. If several parents have an attribute with the same name, Python takes it from the parent that is written **first** (on the left):

````python
class LeftBase:
    a = "a left"
    b = "b left"

class MiddleBase:
    b = "b middle"
    c = "c middle"

class RightBase:
    c = "c right"
    d = "d right"

class DerivedClass(LeftBase, MiddleBase, RightBase):
    pass

for attr in "abcd":
    print(f"{attr}: {getattr(DerivedClass, attr)}")
````

Multiple inheritance can create a **diamond**. This structure is possible, but it is better to avoid it:

````python
#      Animal
#    /       \
#  Cat       Dog
#    \       /
#      CatDog
````

**Diamond inheritance** is a situation in languages with multiple inheritance in which two classes `B` and `C` inherit from `A`, and the class `D` inherits from both `B` and `C`. This structure can be ambiguous. Imagine that an object of the class `D` calls a method that is defined in the class `A`, and the classes `B` and `C` have overridden this method in different ways. From which class should `D` take the method: from `B` or from `C`?

In Python, this problem became important when all classes received the common parent `object`: since then, almost every case of multiple inheritance contains a diamond. To solve the problem, Python builds a list of classes in a fixed order, the **MRO** (Method Resolution Order). Since version 2.3, Python uses the **C3 linearization** algorithm to build this list.

````python
#      A
#   /    \
#  B      C
#   \   /
#     D
````

For the diamond, we can explain the result in a simplified way. First, we go through the classes in depth, starting from the left: $(D, B, A, C, A)$. Then we remove every repeated class except its last occurrence. So the final order is $[D, B, C, A]$ (and `object` at the end).

As a result, Python moves level by level: it never goes to a parent class before it has visited all its children, even if this parent has several children.

````python
class CatDog(Cat, Dog):
    def __init__(self):
        super().__init__()
        print("i am a CatDog!")
````

````python
catdog = CatDog()
catdog.some_value
````

The order of the parent classes is important:

````python
class DogCat(Dog, Cat):  # now the other way round
    def __init__(self):
        super().__init__()#The super() function returns a proxy object
        #that passes method calls to a parent or sibling class of the given type
        print("i am a DogCat!")

dogcat = DogCat()
dogcat.some_value
````

What about methods? Python searches for them in the same way as for other attributes.

As we said above, Python searches for methods and attributes in the following order:

1) in the object (in its `__dict__`);

2) in the class of the object;

3) in the parents of the class, in the order of the MRO.

The attribute `__mro__` shows this order:

### `super()` and the diamond

Now we can understand why `super()` is better than a direct call of the parent method. Let us build a diamond and call the initializers of the parent classes directly by their names:

````python
class North:
    def __init__(self):
        print("North")

class West(North):
    def __init__(self):
        print("West")
        North.__init__(self)

class East(North):
    def __init__(self):
        print("East")
        North.__init__(self)

class South(West, East):
    def __init__(self):
        print("South")
        West.__init__(self)
        East.__init__(self)

s = South()
````

The initializer of `North` runs **twice**. If it opened a file or added money to an account, this would be a serious error. Now let us use `super()` in all the classes:

````python
class North:
    def __init__(self):
        print("North")

class West(North):
    def __init__(self):
        print("West")
        super().__init__()

class East(North):
    def __init__(self):
        print("East")
        super().__init__()

class South(West, East):
    def __init__(self):
        print("South")
        super().__init__()

s = South()
````

Now each initializer runs only once. Note that `super()` in the class `West` calls the method of `East`, not the method of `North`! This is because `super()` does not simply mean "my parent". It means "**the next class in the MRO** of the current object". For an object of the class `South`, the MRO is the following:

````python
print([cls.__name__ for cls in South.__mro__])
````

This works correctly only if **all** the classes in the hierarchy use `super()`.

### Mixins: multiple inheritance done right

A **mixin** is a class that adds some functionality to other classes, but it is not used on its own. The main purpose of a mixin is to give extra methods to its child classes.

In other words, mixins are a limited form of multiple inheritance. In Python, a mixin is a parent class that gives some functionality to its child classes, but we do not create instances of the mixin itself. In addition, it is better if mixins do not inherit from other mixins and do not have their own state (attributes). The names of mixin classes usually end with `Mixin`.

````python
class MusicPlayerMixin:
    def play_music(self, song):
        print("Now playing: {}".format(song))
````

````python
class Auto:
    def ride(self):
        print("Riding on a ground")

class Boat:
    def swim(self):
        print("Sailing in the ocean")

class Amphibian(Auto, Boat,MusicPlayerMixin):
    def __init__(self):
        super().__init__()
````

````python
a = Amphibian()
a.ride()
a.swim()
a.play_music('hi')
````

A mixin can also use the attributes of the class to which it is added. For example, the following mixin describes any object in a readable form:

````python
class DescribeMixin:
    def describe(self):
        params = ', '.join(f'{key}={value!r}' for key, value in self.__dict__.items())
        return f'{type(self).__name__}({params})'


class Book(DescribeMixin):
    def __init__(self, title, author):
        self.title = title
        self.author = author


class Student(DescribeMixin):
    def __init__(self, name, group):
        self.name = name
        self.group = group


print(Book('War and Peace', 'Leo Tolstoy').describe())
print(Student('Anna', 'B-101').describe())
````

### Abstract classes

Look again at the class `Animal`. Its method `speak` only raises `NotImplementedError`. In fact, `Animal` describes an **interface**: every animal must be able to speak, but each child class does it in its own way. However, nothing stops us from creating the object `Animal()` or a child class without the method `speak`. We see the error only when we call the method.

The module `abc` (Abstract Base Classes) solves this problem:
- an **abstract class** inherits from the class `ABC`;
- an **abstract method** is marked with the decorator `@abstractmethod`;
- we **cannot create an instance** of a class if at least one of its abstract methods is not implemented.

````python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def perimeter(self):
        raise NotImplementedError

    @abstractmethod
    def area(self):
        raise NotImplementedError


try:
    Shape()
except TypeError as msg:
    print(msg)
````

Let us create a child class that implements only one of the two abstract methods:

````python
from math import pi


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def perimeter(self):
        return 2 * pi * self.radius


try:
    Circle(1)
except TypeError as msg:
    print(msg)
````

The error message shows which methods are not implemented yet. Now let us implement all the methods:

````python
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def perimeter(self):
        return 2 * pi * self.radius

    def area(self):
        return pi * self.radius ** 2


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def perimeter(self):
        return 4 * self.side

    def area(self):
        return self.side ** 2


c = Circle(1)
print(c.area(), c.perimeter())
print(isinstance(c, Shape))
````

So an abstract class is a **contract**: if a class inherits from `Shape`, it must implement `area` and `perimeter`. Python checks this when we create an object, not when we call a method.

**Summary:**
- `class Child(Parent)` creates a child class; every class inherits from `object`.
- Python searches for an attribute in the object, then in its class, then in the parent classes in the order of the MRO (`Class.__mro__`).
- A child class can inherit, override or extend the methods of the parent class.
- `super()` calls the method of the next class in the MRO; we do not pass `self` to it.
- `isinstance(obj, Class)` takes inheritance into account; `type(obj) == Class` does not.
- Protected attributes (`_name`) are available in child classes; private attributes (`__name`) are not, because of name mangling.
- A child class can override a protected method, but it cannot override a private method.
- With multiple inheritance, it is better to use mixins and to avoid diamonds.
- An abstract class (`ABC` + `@abstractmethod`) describes an interface; we cannot create its instances.

## Polymorphism

**Polymorphism** is the ability of code to work in the same way with objects of different types. In Python, there are three main forms of polymorphism:
1. **subtype polymorphism** — code works with a parent class, and each child class gives its own implementation of a method;
2. **parametric polymorphism and duck typing** — code works with any object that supports the necessary operations;
3. **ad-hoc polymorphism** — one function name has different implementations for different types of arguments.

In class, we discuss the first two forms, because they are closely connected with inheritance. Sections 3–5 are for self-study.

### 1. Subtype polymorphism

The function `print_area` works with any shape. It does not know which method `area` it calls: Python chooses the method at run time, according to the real class of the object.

````python
def print_area(shape):
    print(f'{type(shape).__name__}: area = {shape.area():.2f}')


shapes = [Circle(1), Square(2), Circle(0.5)]
for shape in shapes:
    print_area(shape)
````

The same idea works with our employees. We call the method `work()` for every employee, and each object uses its own version of the method:

````python
staff = [Employee('Ivan Sidorov', 2500), Programmer('Vasya Petrov', 3000, 'Python')]
for person in staff:
    print(person.work())
````

If we add a new class, for example `Triangle(Shape)` or `Tester(Employee)`, this code works with it without any changes. This is the main advantage of polymorphism: the code is easy to extend.

### 2. Parametric polymorphism and duck typing

In Python, inheritance is **not necessary** for polymorphism. A function works with any object that has the necessary methods. This idea is called **duck typing**: *"If it walks like a duck and quacks like a duck, then it is a duck."* Python does not check the type of an object; it only checks whether the object can do what we ask.

For example, the function `add` works with numbers, strings and lists, because all these types support the operator `+`:

````python
def add(x, y):
    return x + y


print(add(2, 2))
print(add(3.14, 2.71))
print(add('Hello, ', 'World!'))
print(add([1, 2], [3]))
````

In the same way, the following classes have no common parent, but the function `make_it_quack` works with both of them:

````python
class Duck:
    def quack(self):
        return 'Quack!'


class Robot:
    def quack(self):
        return 'Beep... quack!'


def make_it_quack(thing):
    print(thing.quack())


for thing in [Duck(), Robot()]:
    make_it_quack(thing)
````

Many built-in functions work in the same way. For example, `len()` works with any object that has the method `__len__`, and the loop `for` works with any object that can be iterated. We will discuss such special methods below.

### 3. Ad-hoc polymorphism: `isinstance` and `singledispatch` *(self-study)*

In some languages (for example, C++ or Java), we can write several functions with the same name and different types of parameters. This is called **function overloading**. Python does not support overloading: a new `def` with the same name simply replaces the old function:

````python
def double_it(x: int):
    return 2 * x


def double_it(x: list):          # this function replaces the first one
    return [2 * value for value in x]


print(double_it([1, 2, 3]))
print(double_it(5))
````

One solution is to check the type of the argument with `isinstance`:

````python
def double_it(x):
    if isinstance(x, int):
        return 2 * x
    if isinstance(x, list):
        return [double_it(value) for value in x]
    raise TypeError(f'double_it does not support the type {type(x).__name__}')


print(double_it(1))
print(double_it([1, 2, 3]))
````

Another solution is the decorator `singledispatch` from the module `functools`. It chooses the implementation according to the type of the **first** argument:
- the main (generic) function is marked with `@singledispatch`; it runs when no other implementation is suitable;
- the implementations for particular types are registered with `@<function>.register(<type>)`; their names are usually `_`.

````python
from functools import singledispatch


@singledispatch
def double_it(x):
    # the generic function: it runs if no registered type is suitable
    raise TypeError(f'double_it does not support the type {type(x).__name__}')


@double_it.register(int)
def _(x):
    return 2 * x


@double_it.register(list)
def _(x):
    return [double_it(value) for value in x]


print(double_it(1))
print(double_it([1, 2, 3]))
````

````python
print(double_it.registry.keys())

try:
    double_it(1.5)
except TypeError as msg:
    print(msg)
````

### 4. Abstract classes of the standard library

The function `double_it` works only with `int` and `list`. However, `float`, `complex` and `Fraction` are numbers too, and a tuple or a set can also contain numbers. The standard library has abstract classes for such general types: `numbers.Number` (all numbers) and `collections.abc.Collection` (all collections). Built-in types are registered as their subclasses, so `isinstance` and `issubclass` work with them:

````python
from numbers import Number
from collections.abc import Collection
from fractions import Fraction
from decimal import Decimal

for t in [bool, int, Fraction, Decimal, float, complex]:
    print(f'{t.__name__} is a Number: {issubclass(t, Number)}')

for t in [tuple, list, str, dict, set, int]:
    print(f'{t.__name__} is a Collection: {issubclass(t, Collection)}')
````

````python
def double_it(x):
    if isinstance(x, Number):
        return 2 * x
    if isinstance(x, Collection) and not isinstance(x, str):
        return [double_it(value) for value in x]
    raise TypeError(f'double_it does not support the type {type(x).__name__}')


print(double_it([1, 2., 3. + 4.j, Fraction(5, 6), {9., 10.}]))
````

Note that a string is also a collection, so we exclude it explicitly. Otherwise, the function would call itself again and again for a string of one character, because such a string contains itself.

### 5. Polymorphism of operators: special methods *(self-study)*

Operators and built-in functions are polymorphic too. The operator `+` adds numbers, joins strings and joins lists. How does it work? For each operator, Python calls a **special method** of the object. Such methods are also called *magic methods* or *dunder methods* (from "**d**ouble **under**score"):

| Code | What Python calls |
|---|---|
| `print(obj)`, `str(obj)` | `obj.__str__()` |
| `len(obj)` | `obj.__len__()` |
| `abs(obj)` | `obj.__abs__()` |
| `a + b`, `a - b`, `a * b` | `a.__add__(b)`, `a.__sub__(b)`, `a.__mul__(b)` |
| `a == b` | `a.__eq__(b)` |
| `a < b`, `a <= b`, `a > b`, `a >= b` | `a.__lt__(b)`, `a.__le__(b)`, `a.__gt__(b)`, `a.__ge__(b)` |

If we define these methods in our class, its objects can work with operators and built-in functions:

````python
class Money:
    def __init__(self, amount, currency='RUB'):
        self.amount = amount
        self.currency = currency

    def __str__(self):
        return f'{self.amount} {self.currency}'

    def __add__(self, other):
        if self.currency != other.currency:
            raise ValueError('The currencies are different')
        return Money(self.amount + other.amount, self.currency)

    def __eq__(self, other):
        return self.amount == other.amount and self.currency == other.currency

    def __lt__(self, other):
        return self.amount < other.amount


a = Money(100)
b = Money(250)
print(a + b)
print(a == Money(100), a < b, b > a)
print(sorted([b, a, Money(50)])[0])
````

Note that `b > a` also works, although we have not defined `__gt__`. If the left object cannot perform the comparison, Python tries the "mirror" operation `a < b`. If we need all six comparison operators, we can define only `__eq__` and `__lt__` and add the decorator `@functools.total_ordering` to the class: Python creates the other operators automatically.
