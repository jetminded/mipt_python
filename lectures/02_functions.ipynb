## Basic syntax

In Python, you declare a function like this:

```python
def function_name(args):
    do_something()
    return something_else
```

````python
def function():
    pass
````

**The keyword `pass` is useful when you need to show clearly that a block of code does nothing.**

**Note 1.** In Python, a function always *returns* a value. If there is no `return` in the body of the function, it returns the special value `None`.

**Note 2.** Python does not have strict typing. It does not check the types of the arguments, and it does not check whether you can do an operation with them. Be careful and check what you pass to a function!

Let's create a very simple function that returns the quotient of its two arguments (the first argument divided by the second):

````python
def division(a, b):
    return a / b

result = division(4, 2)
print(result, type(result))
````

````python
# A similar function that returns "nothing"
def print_division(a, b):
    print("Result:", a / b)

result = print_division(4, 2)
print(result, type(result))
````

**Note:** A function in Python can return several values. In this case, Python automatically puts these values into a tuple and returns this tuple:

````python
def divide_and_return_args(a, b):
    return a / b, a, b
````

````python
result = divide_and_return_args(4, 2)
result1, result2, result3 = divide_and_return_args(4, 2)
print('A tuple returned from the function:', result)
print('Unpacked tuple:', result1, result2, result3)
````

## Mutable and immutable arguments

How Python passes arguments to a function:

>- Mutable arguments (lists, dictionaries, sets) behave as if they are passed by reference: if the function changes them, you can see the changes outside the function.
>- Immutable arguments (numbers, strings, tuples) behave as if they are passed by value: the function cannot change the original object.

Let's look at an example. The function takes 2 arguments: the dictionary `team` (a mutable argument) and the integer `number` (an immutable argument). The function changes both arguments inside its body:

````python
def get_my_hero_team(team: dict, number: int):
    number = 10
    team['Chuck'] = 'Norris'
    team['Sylvester'] = 'Stallone'
````

Let's see what happens to our variables outside the function. Will they change after we call the function?

````python
number = 5
hero_team = {'Bruce': 'Willis', 'Chuck': 'Lorre'}
get_my_hero_team(hero_team, number)
````

The variable `number` has not changed its value:

````python
number
````

The variable `hero_team` has changed its value:

````python
hero_team
````

Why did this happen? In Python, numbers (`int`, `float`), strings, tuples, boolean values (`True`, `False`) and bytes are **immutable**. You cannot change an object of these types: when you "change" it, Python creates a new object. So the line `number = 10` inside the function does not change the old object. It only makes the local name `number` refer to a new object, and the variable `number` outside the function stays the same.

A dictionary is **mutable**. The function gets the same dictionary object, not a copy, so all the changes are visible outside the function.

## Positional and keyword arguments

You can pass arguments to a function in two ways:

- **by position** — you simply list the values of the arguments. The first value in the function call goes to the first argument in the function definition, the second value goes to the second argument, and so on:

````python
def simple_division(a , b):
    print(f"Function received {a=}, {b=}")
    return a / b
````

````python
simple_division(4, 2)  # When we call the function, a gets 4 and b gets 2
````

- **by keyword** — you clearly say which argument gets which value:

````python
simple_division(b=2, a=4)  # You can list keyword arguments in any order
````

**PEP 8:** When you pass keyword arguments, do not put spaces around the `=` sign!

You can mix both ways. For example, you can pass `a` by position and `b` by keyword:

````python
simple_division(4, b=2)
````

**But** remember the rule: positional arguments go first, and keyword arguments go after them:

````python
simple_division(b=2, 4)
````

````python
simple_division(2, a=4)
````

````python
simple_division(a=4)
````

In Python, everything is an object: ordinary variables, functions, modules... So if you write the name of a function without calling it (without parentheses and arguments), Python treats it as a separate object (a "variable") with its own address in memory and its own name:

````python
simple_division
````

## Default arguments

You can give an argument a default value. If you do not pass a value for this argument when you call the function, Python uses the default value:

````python
def safe_division(a, b, check_zero=True):
    """
    This function returns division of two arguments and checks whether the divisor is equal to zero
    """
    if check_zero and b == 0:
        print('Caught division by zero!')
        return None
    else:
        return a / b
````

**Note:** Like in many other languages, a function in Python can have more than one `return`.

````python
safe_division(4, 0)
````

````python
safe_division(4, 0, check_zero=False)
````

## Starred expressions

When you call a function in Python, you can pass not only a fixed number of arguments, but also a variable number of them.

For this, we use the **unpacking operators** (*starred expressions*) `*` and `**`. The operator `*` unpacks a list or a tuple, and the operator `**` unpacks a dictionary. Let's look at an example:

````python
print(1,2,4,3,3,3)
````

````python
test_list = ['1', 1, False, [3], -5+1j]

print(test_list)

print(*test_list)
print(test_list[0], test_list[1], test_list[2], test_list[3], test_list[4])
````

So the unpacking operator `*` turns a container into a set of positional arguments.

The unpacking operator `**` works in a similar way with a dictionary, but it turns the dictionary into a set of keyword arguments.

````python
test_dict = {"sep": "\n", 'end': 'END OF PRINT'}
# We pass "\n" as the separator and "END OF PRINT" as the end of the line
print(**test_dict)

print()
print()

print("Hello, World!", "This is my first program!", sep="\n", end='END OF PRINT')
````

You can also use the unpacking operator when the number of values is unknown or can change:

````python
test_list = [1, 1, 2, 3, 5, 8, 13]

first_value, second_value, *other_values = test_list
print(first_value, second_value, other_values)
````

The variable `other_values` can also get 0 elements:

````python
shorter_list = test_list[:1]

first_value, *other_values = shorter_list
print(first_value, other_values)
````

However, all the other variables must get a value:

````python
test_list_shortest = test_list[:1]

first_value, second_value, *other_values = test_list_shortest
print(first_value, second_value, other_values)
````

In the same way, if you forget the star, Python thinks that you want to assign exact values, not a list of "all the other values". Then the number of values must be the same as the number of variables. If there are fewer or more values, Python raises an error. As people say, it is easier to show than to explain:

````python
first_value, second_value, other_values = test_list
print(first_value, second_value, other_values)
````

**Life hack**

To ignore extra returned values, you can use the popular symbol `_`. In fact, this creates a normal variable with the name `_`, and you can use it later in the code. But usually programmers use this name for values that they do not need:

````python
first_value, *_ = test_list
print(first_value, _)
````

## Variable number of arguments

You can also use the unpacking operators when you declare a function. Then the function can take any number of arguments. In general, a function declaration looks like this:

```python
def foo(
    positional or keyword args,  # arguments that you can pass both by position and by keyword
    *[other positional args],    # usually *args: a tuple with any number of extra positional arguments (a bare * only shows where keyword-only arguments start)
    keyword-only args,           # keyword arguments that you cannot pass by position
    **[other keyword args]       # usually **kwargs: a dictionary with any number of extra keyword arguments
)
```

Of course, you do not have to use all these kinds of arguments. You can use only the ones you need.

We can declare a function that takes a variable number of positional arguments. For example, we can use it to join several lists into one list (the function `chain` from the module `itertools` does the same):

````python
def merge_lists(*lists):
    print("Input elements:", lists)
    main_list = []
    for lst in lists:
        main_list.extend(lst)
    return main_list

print(merge_lists([1], ['a', 'b']))
print(merge_lists([1], ['a', 'b'], [True, False]))
print(merge_lists())
````

The operator `**` works in the same way with keyword arguments:

````python
def print_keyword_args(**kwargs):
    print("Input kwargs:", kwargs)
    print("Answer:", kwargs.get("answer", "Not Found"))

print_keyword_args(a="1", b=[2], answer=42)
````

So we can declare a function that takes any number of any arguments:

````python
def uberfunc(*args, **kwargs):
    print('args =', args, 'kwargs =', kwargs)

uberfunc()
uberfunc(1, 2, 3, a=4)
uberfunc(answer=42)
````

**Note**

Usually we pass the most important arguments to a function by position, and we use keyword arguments to change small details of how the function works.

**Note for advanced students**

The construction `(*args, **kwargs)` is often used with class inheritance. It helps to initialize the parent class with parameters that the child class may not use. For example:

```python
class Son(Parent):

    def __init__(self, specific_argument, *args, **kwargs):
        super().__init__(*args, **kwargs)  # initialize the parent class and pass all the "extra" arguments to it
        do_something(specific_argument)
```

Some libraries use the same idea inside their functions: they pass arguments to "lower-level" functions.

## Anonymous functions

An **anonymous function** is a special kind of function. You declare it in the place where you use it, and it does not get a name. In Python, anonymous functions are called **lambda functions**. A lambda function can take any number of arguments, but it can have only one expression (one line). When you call the function, it returns the result of this expression.

The basic syntax of a lambda function is:

```python
lambda x: expression
```

Let's create an anonymous function and save it in a variable:

````python
f_lambda = lambda x: print(x, end='EOL')
````

Let's check that we really created a function:

````python
type(f_lambda)
````

Let's check how our lambda function works:

````python
f_lambda("Hello!")
````

We saved our lambda function in a variable, but in real code people usually do not do this. If you plan to use a function more than once, it is always better to declare it with `def ...`.

The function that we created works exactly like this ordinary (named) function:

````python
def f(x):
    return print(x, end='EOL')

f("Hello!")
````

Like an ordinary function, a lambda function can take several arguments and have default values:

````python
print((lambda x, y=4: x + y)(10, 13000))
print((lambda x, y=4: x + y)(10))
print((lambda x, y=4, *args: x + y + sum(args))(10, 13000, 4, 100))
````

### Using lambda functions

**Why do we need lambda functions?** Some functions take as an argument not an ordinary variable, like a number or a string, but another function. They use this function for some calculations inside, and usually this function is quite simple.

A good example is the function `sorted`. It sorts containers in ascending order. By default, it compares the elements of the container with each other.

````python
test_list = [3, 14, 15, 2, 73]

sorted(test_list)
````

Sometimes the elements of the container that we want to sort are also containers, for example, tuples or lists. In such cases we often do not want the standard lexicographic sorting. Instead, we want to sort by one particular element of each inner container.

Let's look at an example. We have a list `customer_data`. It contains pairs of elements (tuples of two elements): "customer name" and "income received".

````python
customer_data = [("ivan", 2), ("george", 4), ("carl", 1), ("liz", 20), ("vlad", 10)]
````

Let's try to sort this list in the standard way:

````python
sorted(customer_data)
````

Python sorted the list by the first element (lexicographically, as expected). But we want to sort the list by the income from each customer.

The optional argument `key` of the function `sorted` can help us here. It takes a function (do not confuse it with the value or the result of a function). This function takes one argument — an element of the list — and returns one value. Python sorts the list by these values.

When we need a function only once for a small task, like here, we use a lambda function. Let's sort the list with the argument `key`:

````python
sorted(customer_data, key=lambda x: x[1])  # We sort by the element with index 1
````

As you can see, the list is now sorted by the element with index 1. But what if we want to sort it in descending order? Another argument can help us here — `reverse`. It sorts the list in the reverse (descending) order:

````python
sorted(customer_data, key=lambda x: x[1], reverse=True)
````

Another example of a built-in function that takes another function as an argument is the function `map`.

The function `map` applies one function to all the elements of a container. It returns the result as a special object (an iterator), so we convert it to a list.

````python
list(map(lambda x: x ** 2, [1, 2, 3]))
````

Now we only need to convert the elements from strings to integers:

## Function documentation and annotations

````python
import typing as tp

def relative_difference(x: float, y: float) -> float | None:
    """
    Returns the relative difference between two numbers
    The second line is here only to show an example of a multi-line docstring
    """
    delta = x - y
    mean = (x + y) / 2
    if mean == 0.0:
        return None
    return abs(delta / mean)
````

## Recursion

A function is **recursive** if it calls itself while it runs. Let's look at recursion using one of the simplest sorting algorithms as an example.

The main idea of the algorithm:
1. Current list = input list
2. Find the minimum element in the current list
3. Remove the minimum element from the current list — we get a new list
4. Repeat steps 2–3 for the new list, then `sorted_input_list = [min_elem] + sorted_current_list`

````python
def easy_sort(x):
    if len(x) == 1:
         return x  # when Python runs return, the function returns the value and stops
    first = min(x)
    x.remove(first)
    print("Min:", first, "rest:", x)
    result = [first] + easy_sort(x)
    print("Result:", result)
    return result
````

````python
easy_sort([4, 2, 3, 1, 7, 5])
````

**How does recursion work?**

Before Python calls the function again, it remembers the current state of the previous (parent) call and puts it on the call stack in the computer's memory.

This is the danger of recursion: memory is limited, so the depth of recursion is limited too (how many times you can call a function inside a function inside a function...).

Every recursive function must have a stopping condition. Without it, the function will call itself forever.

Python has a strict limit on the depth of recursion, so that a program does not fill up all the memory:

````python
import sys
sys.getrecursionlimit()
````

You can change this limit like this:

````python
sys.setrecursionlimit(1_000_000)
````

but we do not recommend doing this!

The limit is small on purpose: your program should crash long before it uses all the RAM. Otherwise, other programs on the same computer can also start to crash.

For this reason, it is better to replace recursive functions with loops when it is possible.
