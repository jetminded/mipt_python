## Errors and exceptions

Python has at least two kinds of errors: **syntax errors** and [**exceptions**](https://docs.python.org/3/tutorial/errors.html).

- A **syntax error** means that the code breaks the rules of the Python language. Python cannot even start such a program. Beginners see these errors very often.
- An **exception** is an error that Python finds **while the program is running**. The code is written correctly, but something goes wrong during the work.

An exception does not always crash the program. We can **handle** it (react to it), and then the program continues to work normally.

But if nobody handles the exception, Python stops the program and prints an error message.

Examples of exceptions:

````python
print(1 / 0)
````

````python
open('nonexistent.file')
````

### How to read an error message

Look at the messages above.

- The **last line** says what happened.
- Exceptions have different **types**. The type is written before the colon. In the first example it is [`ZeroDivisionError`](https://docs.python.org/3/library/exceptions.html#ZeroDivisionError).
- After the type there is extra information. It depends on the type of the error and on its cause.
- Everything above the last line is the **traceback**. It shows the place in the code where the error happened.

The traceback is easier to understand when the error happens inside several functions:

````python
def f():
    1 / 0

def g():
    f()

g()
````

This time the message is longer, because the error did not happen at the top level of the program. It happened inside the function `f`, and `f` was called from the function `g`. The traceback shows the whole path: the call `g()`, then the call `f()`, then the line with the error.

### Built-in exceptions

Python has many built-in exceptions. The interpreter raises them when it finds an error at run time. Here are the ones you will see most often:

| Exception | When it happens | Example |
|---|---|---|
| [`NameError`](https://docs.python.org/3/library/exceptions.html#NameError) | you use a name that does not exist | `print(x)` before `x` is created |
| [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError) | an operation gets an object of the wrong type | `'abc' + 1` |
| [`ValueError`](https://docs.python.org/3/library/exceptions.html#ValueError) | the type is right, but the value is wrong | `int('abc')` |
| [`IndexError`](https://docs.python.org/3/library/exceptions.html#IndexError) | the index is outside the list | `[1, 2, 3][10]` |
| [`KeyError`](https://docs.python.org/3/library/exceptions.html#KeyError) | the key is not in the dictionary | `{'a': 1}['b']` |
| [`AttributeError`](https://docs.python.org/3/library/exceptions.html#AttributeError) | the object does not have this attribute or method | `'abc'.push('d')` |
| [`ZeroDivisionError`](https://docs.python.org/3/library/exceptions.html#ZeroDivisionError) | division by zero | `1 / 0` |
| [`FileNotFoundError`](https://docs.python.org/3/library/exceptions.html#FileNotFoundError) | the file does not exist | `open('nonexistent.file')` |

Sometimes Python raises an exception **not** because of an error, but as a signal. For example, an iterator raises [`StopIteration`](https://docs.python.org/3/library/exceptions.html#StopIteration) to say that it has no more elements. A `for` loop waits for this signal and then stops. So `StopIteration` is not a mistake in your code. It is just a way to send a message.

In Python everything is an object, and exceptions are objects too. Every exception has a type: `NameError`, `KeyError`, `StopIteration` and so on. When Python finds an error, it creates an exception object of the right type. Then this exception starts to **propagate**: it moves up through the program until somebody handles it.

## Handling exceptions: `try` / `except`

We handle errors with the `try` statement. It has two main forms:

- `try/except`;
- `try/finally`.

With `try/except` we can handle exceptions of a specific type. The simplest form looks like this:

```python
try:
    statements
except ExceptionType:
    statements
```

After the word `try` we put the code that can raise an exception. The code in the `except` block runs only if:

1. an error happens while the `try` block is running, **and**
2. the type of this error matches `ExceptionType`.

Let's divide by zero again. This time we put the division inside `try/except` and handle `ZeroDivisionError`.

````python
try:
    1 / 0
    print('After the division.')
except ZeroDivisionError:
    print('Division by zero was caught!')
````

We can see that:

1. There is no error message and no traceback, because the error was handled.
2. The `print` in the `except ZeroDivisionError` block ran instead.
3. The `print` in the `try` block after the division did **not** run. The error happened earlier and started to propagate.

When an exception appears, the normal work of the program stops. The exception moves up through the chain of function calls (the **call stack**). When it meets the first `except` block with a matching type, it stops there, and the code in that block runs.

Here are a few examples of how this works.

The first example shows that the exception keeps moving until it meets the **right** `except` block.

````python
try:
    try:
        1 / 0
    except NameError:
        print('The error was caught by the inner try/except block.')
except ZeroDivisionError:
    print('The error was caught by the outer try/except block.')

print('A statement after the exception was handled.')
````

Here `ZeroDivisionError` appeared in the inner `try/except` block. But the inner block only catches `NameError`, so the types did not match. The inner block did not handle the error, and the exception moved on to the outer block. The outer block catches `ZeroDivisionError`, so the exception stopped there and the outer `print` ran. After that the program continues in the normal way.

The next example shows that the exception stops at the **first** matching `except` block.

````python
try:
    try:
        1 / 0
    except ZeroDivisionError:
        print('The error was caught by the inner except block.')
except ZeroDivisionError:
    print('The error was caught by the outer except block.')

print('A statement after the exception was handled.')
````

Now the inner block catches the right type. The exception stops in the inner block, and the inner `print` runs. For the outer block nothing bad happened: the error was handled before it could get there. So the outer `except` is skipped.

One `try` can have several `except` blocks. This way we can react to different types of errors in different ways.

Handling exceptions: `try...except...except`

````python
path = 'missing_data.txt'

try:
    file_obj = open(path, 'r')
except FileNotFoundError:
    print(f'File {path!r} was not found')
except (TypeError, ValueError, OSError) as err:
    print('Demonstrating handling multiple exception types in one clause')
except Exception as err:  # the first matching except block is executed
    print(f'An error occurred while opening {path!r}: {err!r}')
````

What is new in this example:

- **Several `except` blocks.** Python reads them from top to bottom and runs the **first** block that matches. All other blocks are skipped.
- **A tuple of types.** `except (TypeError, ValueError, OSError)` catches an exception of any of these types.
- **`as err`.** This gives us the exception object, so we can work with it:

```python
except ExceptionType as target:
    do something with target
```

Some exception objects have extra information about what happened. All standard exceptions can be turned into a string: the string is the message of the exception.

````python
try:
    int('abc')
except ValueError as err:
    print(err)
    print(repr(err))
    print(type(err))
````

### The order of `except` blocks

Python reads the `except` blocks from top to bottom and stops at the first one that matches. Exception types form a hierarchy, so there is a rule:

> **Tip.** Always handle the most specific exceptions first, and the more general ones after them.

If you do not follow this rule, the specific `except` blocks will never run. Let's check. `Exception` is the parent of almost all built-in exceptions, so it is one of the most general types.

````python
try:
    1 / 0
except Exception:
    print('An unexpected exception was caught!')
except ZeroDivisionError:
    print('ZeroDivisionError was caught!')
````

Here the `try` block raises `ZeroDivisionError`. Python checks `except Exception` first. It matches, because `ZeroDivisionError` is a kind of `Exception` (`Exception` $\to$ `ArithmeticError` $\to$ `ZeroDivisionError`). So this block runs, and Python never gets to `except ZeroDivisionError`: the error is already handled.

The correct way is to swap the blocks.

````python
try:
    1 / 0
except ZeroDivisionError:
    print('ZeroDivisionError was caught!')
except Exception:
    print('An unexpected exception was caught!')
````

## `else` and `finally`

A `try` statement can have two more blocks:

- `else` runs only if the `try` block finished **without** errors;
- `finally` runs **always**: when there was no error, when the error was handled, and when nobody handled the error.

### The `else` block

We write `else` after all `except` blocks. As an example, here is a function `divide`. It tries to divide the first argument by the second one. If the division works, it returns the result. If the second argument is zero, it returns infinity.

````python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print('Tried to divide by zero.')
        return float('inf')
    else:
        print('The division was successful.')
        return result
````

The `else` block in this function runs only if the division in the `try` block works, that is, if there is no `ZeroDivisionError`.

````python
print(divide(42, 14))
print(divide(42, 0))
````

### The `try/finally` block

```python
try:
    statements
finally:
    statements
```

The code in the `finally` block runs in any case: with an error in the `try` block or without it. If there is an error, the `finally` block runs first, and then the error continues to propagate.

People often call this block a **cleanup handler**. We put there the code that cleans up after the work (for example, gives back a resource). We can be sure that this code will run.

For example, in the past people often wrote code like this:

```python
f = open(some_file, 'w')
try:
    do_something_with_file(f)
finally:
    f.close()
```

All the work with the open file is inside `try/finally`. This guarantees that the file is closed even if an exception happens. Today it is much easier to use context managers (`with`) for this. We will talk about them below.

### The `try/except/finally` block

A `try` statement can have both `except` and `finally`. You can think of it as a `try/except` block **inside** a `try/finally` block. These two pieces of code work in the same way:

```python
try:
    statements
except SomeException:
    statements
else:
    statements
finally:
    statements
```

```python
try:
    try:
        statements
    except SomeException:
        statements
    else:
        statements
finally:
    statements
```

Here is an example from the Python documentation.

````python
def divide(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        print('Tried to divide by zero!')
    else:
        print(f'The result is {result}')
    finally:
        print("Running the 'finally' block.")
````

The function `divide` tries to divide `x` by `y`.

If the division works (`else`), the function prints the result.

````python
divide(2, 1)
````

If there is a division by zero, the error is caught and the function prints a message.

````python
divide(2, 0)
````

If there is an error of another type, it continues to propagate.

````python
divide('2', '1')
````

The `finally` block runs in all three cases.

Handling exceptions: `try...except...else...finally`

One more example. The file `data.txt` does not exist, so `open` raises `FileNotFoundError`. The block `except TypeError` does not match this type, and the `else` block is skipped. But `finally` still runs. After that the exception continues to propagate, and we see the error message.

````python
stream = None

try:
    stream = open("data.txt", "r")  # potentially unsafe operation
    result = 10 / 0
except TypeError as err:  # wrong exception type (scope failure)
    print(f'An issue occurred: {err!r}')
else:  # scope success
    print('Operation completed successfully')
finally:  # scope exit
    if stream is not None:
        stream.close()
    print('This message is always displayed')
````

## Raising exceptions: `raise`

Python raises many exceptions itself. But you can also raise an exception with the keyword `raise`

````python
x = -5
if x < 0:
    raise ValueError('Positive integer expected')
````

After the word `raise` we write an exception object. Usually we also give it a message that explains the problem.

When an exception is raised, the normal work of the program stops. So there are two ways to leave a function: with `return` and with `raise`.

````python
def f():
    print('Before the exception!')
    raise Exception('Exception!')
    print('After the exception!')

f()
````

The statement `print('Before the exception!')` ran, but `print('After the exception!')` did not. Right after `raise` the normal work stopped and the exception started to propagate.

The exception moves up the call stack until it meets a block that handles it. If there is no such block, Python prints an error message and the program crashes. (In interactive mode, for example in a Jupyter notebook, Python does not close. It just waits for your next command.)

You can "raise" only `BaseException` or its child:

````python
raise 50
````

There are also chains of exceptions.

`raise NewError(...) from e` means: "the new exception was caused by `e`". Python remembers both exceptions and shows both of them in the traceback.

````python
def ctr(shows, clicks):
    """Returns banner click-through rate"""
    try:
        try:
            return clicks / shows
        except ZeroDivisionError as e:
            raise ValueError('Bad banner') from e
    except ValueError as e:
        raise OSError from e

ctr(0, 1)
````

You can create your own exception classes. It is enough to inherit from `Exception`. It is good practice to inherit all your exceptions from one common parent. Then it is easier to catch them.

````python
class FarmError(Exception):
    pass

class WrongFieldError(FarmError):
    pass

class WrongCropError(FarmError):
    pass

raise WrongCropError
````

## Error handling strategies: LBYL vs EAFP

As we saw, an exception in Python is not always a sign of a mistake in the program. People often use exceptions to make a program simpler and more reliable.

There are two very different ways to deal with possible errors.

**LBYL** means *"Look Before You Leap"*. Before you do an operation, check that nothing can stop it. Code in this style usually has a lot of `if` checks.

**EAFP** means *"It's Easier to Ask Forgiveness than Permission"*. Just try to do the operation in a `try` block and handle possible exceptions in `except` blocks.

In Python, the second way is better in most situations. Let's look at the main problems of the first one.

### Problem 1. Checks make the code harder to read

Checks can hide the main part of the code, the part that works when there are no errors.

For example, we want to safely calculate the function $f(x) = \sqrt{(x-1)(x-2)(x-3)}$. It is defined only for $x \in [1, 2] \cup [3, +\infty)$.

````python
from math import sqrt

def f(x):
    return sqrt((x - 1) * (x - 2) * (x - 3))
````

If we call this function with an argument outside this set, we get a `ValueError`.

````python
f(2.5)
````

Let's write a function that returns $f(x)$ if $x$ is in the set above, and `None` if it is not.

**LBYL**

````python
def safe_f(x):
    if x < 1:
        return None
    if 2 < x < 3:
        return None
    return sqrt((x - 1) * (x - 2) * (x - 3))

print(safe_f(1.5), safe_f(2.5), safe_f(4))
````

When we read this LBYL code, the first thing we see is the checks of the argument `x`. The main code is hidden at the end of the function.

**EAFP**

````python
def safe_f(x):
    try:
        return sqrt((x - 1) * (x - 2) * (x - 3))
    except ValueError:
        return None

print(safe_f(1.5), safe_f(2.5), safe_f(4))
````

In the EAFP style the main code is at the top of the function, and the special cases are below it.

### Problem 2. Checks repeat the work

A check often does the same work that the operation does anyway.

As an example, take the dictionary method `get`. It returns the value for a key if the key is in the dictionary, and `None` if it is not. Here are two ways to write it.

**LBYL**

````python
def get(d, k):
    if k in d:
        return d[k]
    else:
        return None

print(get({'a': 1}, 'a'), get({'a': 1}, 'b'))
````

Both `k in d` and `d[k]` look for the key in the dictionary. So Python looks for the key two times.

**EAFP**

````python
def get(d, k):
    try:
        return d[k]
    except KeyError:
        return None

print(get({'a': 1}, 'a'), get({'a': 1}, 'b'))
````

Here Python looks for the key only once, in the operation `d[k]`.

### Problem 3. Things can change between the check and the operation

- **Programs with several threads** (several parts of the program run at the same time). Take the dictionary example again. With LBYL, another thread can delete the key right after the check `k in d`. Then `d[k]` raises `KeyError` anyway. EAFP does not have this problem, because there is no separate check.
- **Programs that work with a network.** It is not possible to check in advance that a network operation will finish well. The program does not control everything: the network can fail at any moment. In such cases EAFP is almost the only choice.
- **Files.** The same can happen with files: a file can be deleted by another program after you checked that it exists, but before you opened it.

**In short:** in Python we usually do not ask "can I do this?". We just try, and we are ready to handle the error.

## Context managers

We want to open files and NOT forget to close them.

How can we guarantee that some action is done in any case, whether an exception happened or not?

````python
def do_something_dangerous(fd):
    raise RuntimeError('Not today!')

fd = open('myfile.txt', 'w')
try:
    do_something_dangerous(fd)
finally:
    print('Closing file')
    fd.close()
    print('File closed')
````

Context managers give us a convenient way to set up a "context" and to guarantee that it is cleaned up at the end.

````python
r = acquire_resource()
try:
    use_resource(r)
finally:
    release_resource(r)
````

````python
with acquire_resource() as r:
    use_resource(r)
````

Examples of context managers: `open`

````python
with open('filename.txt', 'w') as fd:
    fd.write("Hello")
# file is closed
fd.write("world")
````

Examples of context managers: `tempfile`

````python
import tempfile

with tempfile.TemporaryFile() as tmp:
    do_something(tmp)
# tmp file is removed
````

Syntax of the `with` statement

````python
# nested contexts
with open('file1.txt') as file1, open('file2.txt') as file2:
    do_something(file1, file2)
# since python 3.10
with (open('file1.txt') as file1, open('file2.txt') as file2):
    do_something(file1, file2)
````

````python
# same as above
with first() as f:
    with second() as s:
        do_something(f, s)
````

````python
with third():  # <as NAME> part is optional
    do_something()
````

Context managers are objects that implement a special protocol

````python
import typing as tp
from types import TracebackType

class MyContextManager:
    def __enter__(self) -> tp.Self: # with () as X
        # initialize context
        return self

    def __exit__(self,
                 exc_type: type[BaseException] | None,
                 exc_value: BaseException | None,
                 traceback: TracebackType | None) -> bool | None:
        # finalize context
        if exc_value is not None:
            return True  # return True from __exit__ to suppress the exception
````

Semantics (what `with` does step by step)

````python
with acquire_resource() as resource:
    use_resource(resource)
````

````python
manager = acquire_resource()
resource = manager.__enter__()
try:
    use_resource(resource)
finally:
    exc_type, exc_value, traceback = sys.exc_info()
    suppress = manager.__exit__(exc_type, exc_value, traceback)
    if exc_value is not None and not suppress:
        raise exc_value
````

A half-joking example

````python
class Tag:
    def __init__(self, name):
        self.name = name
    def __enter__(self):
        print('<{}>'.format(self.name))
        return self

    def __exit__(self, *args):
        print('</{}>'.format(self.name))

    def update(self, new_name):
        self.name = new_name


with Tag('table') as table:
    table.update("broken_table")
    with Tag('tr'):
        with Tag('td'):
            print('cell 1')
        with Tag('td'):
            print('cell 2')
````

`contextlib.contextmanager` is a convenient way to create context managers

````python
from contextlib import contextmanager

@contextmanager
def mycm():
    print('before')
    yield 42  # yep, it is a generator
    print('after')

with mycm() as r:
    print(f'got {r}')

with mycm() as r:
    raise RuntimeError('Oops')
# 'after' is not printed!
````

But you have to be careful with `contextlib.contextmanager`

````python
from contextlib import contextmanager

@contextmanager
def mycm():
    print('before')
    try:
        yield ValueError("kek")
    finally:
        print('after')

with mycm() as r:
    print(type(r), r)
    raise RuntimeError('Oops')
````

Compare the two versions of `mycm`. In the first one, the code after `yield` does not run if an exception happens inside the `with` block. In the second one, `yield` is inside `try/finally`, so the cleanup code runs in any case.
