#  Object-Oriented Programming in Python


# What is OOP?

**A programming paradigm** is a set of ideas and concepts that define the style of writing computer programs (an approach to programming).



**Programming paradigms:**
- **imperative programming** - a sequence of commands that run one after another
- **procedural programming** - we create procedures and functions so that we don't repeat the same code
- **declarative programming** - the focus is on "what" the program must do, not "how" (a bit like SQL)
- **functional programming** - the program is split into functions; the output of one function is the input of another
- **structured programming** - an important step in the history of programming; it gave structure to the main elements of programs: sequence, branching (if/else), the while loop, subroutines (functions)
- **object-oriented programming**

**Object-oriented programming** is a programming paradigm where a program is a set of objects that interact with each other. Each object has its own properties and behavior. Each object is an instance of a certain class, and the classes form an inheritance hierarchy.

**Main terms in OOP:**
- Class
- Objects
- Abstraction
- Encapsulation
- Inheritance
- Polymorphism





## Objects and classes

The main idea of OOP is to find separate entities in a program — objects.

**Examples of objects in real life:** a table, a cup, a person, a book, a building, etc.
**Examples of objects in programs:** a user, a button, a message.

Objects can have their own properties, for example, a color, some content or a user name. These variables (the specific characteristics of an object) are called **attributes**.

Possible actions, for example, "send a message", "hire an employee", "click a button", are called **methods**. They are functions that are described inside an object or a class. They belong to a specific object and let us work with this object or with other parts of the code.

We don't want to write separate methods for every single object (we want to reuse our code), so we have classes.
A **class** is a "template" for an object. It describes the object's attributes and methods. Every object is an instance of some class.

**Example:** we are writing a system to manage employees. Let's create the class "Employee" with these attributes: full name, position, department, manager, salary; and these methods: "move to another department", "raise the salary", "change the position", "fire". One particular employee, Vasya Petrov, is an object — an instance of the class "Employee".

## Abstraction

Abstraction is a way to choose the most important attributes and methods and to leave out the unimportant ones. So abstraction means that we describe an object only with these important characteristics. We should describe an object with the smallest possible set of fields and methods, but the program must still be able to solve its task.

Example: an object of the class "Programmer" probably doesn't need the properties "can cook" or "favorite color". They have nothing to do with programming. But "main programming language" and "work skills" are important properties — you can't describe a programmer without them.

The set of attributes and methods that is available from the outside works as an interface to the object. Other data structures use this interface to work with the object, and they don't need to know how the object works inside.



## Encapsulation

**Every object is an independent structure.** It contains everything it needs to work. If it needs a variable, the variable is described inside the object, not in the outside code. This makes objects more flexible. Even if someone rewrites the outside code, the logic of the object will not change.

Encapsulation lets us hide the details of how a method works from the outside code and from other objects. Other objects don't care (and don't need to know) how exactly an object does its job. The main thing is that it does it :)

So the inside of one object is closed to other objects: from the outside you can only "see" the values of attributes and the results of methods.

## Inheritance

Inheritance lets us create classes and objects that are similar to each other but a little different. For example, they can have extra attributes or other methods. In this case, the more general class becomes the "parent", and the more specific and detailed class becomes the "child".

For example, the class "Employee" can have the child classes "Programmer", "Analyst" and "Tester". Each of them has its own methods, for example, "write code", "analyze", "test", and all the other methods and attributes of the class "Employee" will be in all the child classes. Our Vasya Pupkin will be an instance of both the class "Employee" and the class "Programmer".

One "parent" can have several child classes, and one child class can also have several parents.

Inheritance lets us build complex structures with a clear hierarchy "from general to specific". This makes the code easier to understand and to grow. You don't need to write the same properties again and again in different objects. You just inherit these objects from one "parent", and the parent's properties work in them automatically.

## Polymorphism

The same methods of different classes can do their job in different ways. For example, an "Employee" has the method "work". For a "Programmer", this method means writing code, and for a "Tester" — testing the program. But in general, both of them are work.

Here, a common approach and agreements between developers are important. If a method is called `delete`, it must delete something. How exactly it deletes depends on the object, but the method must do deleting. Moreover, if the team agreed that the "deleting" method is called `delete`, you shouldn't call it `remove` or something else in some object. This makes the code confusing.

## The constructor `__new__`

In OOP theory, a **constructor** is a special block of code that runs when an object is created. When you work with Python, you may think that the method `__init__(self)` is the constructor. But this is not quite true.

In fact, when Python creates an object, it first calls the method `__new__(cls, *args, **kwargs)`, and **this method is the real constructor** of the class:

| Method | Name | What it does |
|---|---|---|
| `__new__(cls[, ...])` | **constructor** | creates a new instance (object) of the class. It gets the class itself as the first argument |
| `__init__(self[, ...])` | **initializer** | gets the new object from the constructor and sets up its attributes |

Please note:
- `__new__()` is called on the class, because the object doesn't exist yet. So its first parameter is `cls` — a reference to the current class (not `self`).
- `__new__()` must **return** the new object. Usually we ask the parent class to create it: `super().__new__(cls)`.
- `__init__()` is the first method that gets the object created by the constructor. It doesn't return anything (it returns `None`).
- Programmers very often override `__init__()` in their classes, because it is a convenient way to set the parameters of a new object when it is created. They override `__new__()` much less often.

Let's see in which order Python calls these methods:

```python
class Phone:
    # Constructor: creates the object
    def __new__(cls, *args, **kwargs):
        print('1. __new__: creating a new object of the class', cls.__name__)
        instance = super().__new__(cls)  # ask the parent class (object) to create the object
        return instance

    # Initializer: sets up the object that __new__ has created
    def __init__(self, color):
        print('2. __init__: setting up the object, color =', color)
        self.color = color
```

```python
my_phone = Phone('Grey')
print(my_phone.color)
```

So `Phone('Grey')` actually does two steps:
1. `Phone.__new__(Phone, 'Grey')` — creates an "empty" object;
2. `__init__(obj, 'Grey')` — fills this object with data.

If `__new__` doesn't return an object of this class, Python does **not** call `__init__`:

```python
class Broken:
    def __new__(cls, *args, **kwargs):
        print('__new__ is called, but it returns nothing')

    def __init__(self):
        print('__init__ is called')  # this line will never run

b = Broken()
print(b)
```

When can `__new__` be useful? For example, when we want to control **how many** objects a class creates. Here is a class that can have only one instance (this pattern is called **Singleton**):

```python
class Settings:
    _instance = None  # here we keep the only object of the class

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            print('Creating the first and only object')
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, theme):
        self.theme = theme
```

```python
s1 = Settings('light')
s2 = Settings('dark')
print(s1 is s2)  # it is the same object
print(s1.theme)  # __init__ was called twice, so the value was overwritten
```

## The destructor `__del__`

An object is created by the constructor `__new__` and set up by the initializer `__init__`. But what happens at the end of an object's life?

When an object is not needed anymore, Python deletes it and frees the memory. This is the job of the **garbage collector**. Just before the object is deleted, Python calls its special method `__del__(self)`. This method is called the **destructor**.

How does Python know that an object is not needed? Python counts the **references** to every object (variables, items in lists, attributes of other objects, and so on). When the number of references becomes 0, nobody can use the object anymore, so Python can delete it.

```python
class Phone:
    def __init__(self, model):
        self.model = model
        print('Phone', self.model, 'is created')

    # Destructor: Python calls it just before the object is deleted
    def __del__(self):
        print('Phone', self.model, 'is deleted')
```

```python
my_phone = Phone('I495')
del my_phone  # we delete the only reference, so the object is deleted too
print('The end of the cell')
```

Important: the `del` statement **does not delete the object**. It deletes only the **name** (the reference). If the object has other references, it stays alive, and `__del__` is not called:

```python
phone1 = Phone('C385')
phone2 = phone1  # now there are two references to one object

del phone1
print('phone1 is deleted, but the object is still alive:', phone2.model)

del phone2  # this was the last reference
```

The destructor is also called when we give a variable a new value and the old object loses its last reference:

```python
phone = Phone('A100')
phone = Phone('B200')  # now the object 'A100' has no references
del phone
```

What if two objects refer to each other? This is called a **reference cycle**. The counters of these objects never become 0, even when we can't reach the objects anymore. Such objects are found and deleted later by a special part of the garbage collector — the module `gc`. We can start it by hand:

```python
import gc

phone_a = Phone('X1')
phone_b = Phone('X2')
phone_a.friend = phone_b  # phone_a refers to phone_b
phone_b.friend = phone_a  # phone_b refers to phone_a: this is a reference cycle

del phone_a, phone_b
print('The names are deleted, but the objects are still in memory')
gc.collect()  # start the garbage collector by hand
print('The garbage collector has done its work')
```

**Be careful with the destructor:**
- You can't know exactly **when** Python will call `__del__`. In the simple examples above it happens at once, but it is not always so. For example, Jupyter Notebook keeps its own hidden references to the results of cells (the variables `_`, `__`, `Out`), so an object can "live" longer than you expect.
- Objects in a reference cycle are deleted only when the garbage collector runs.
- When the program ends, Python doesn't promise to call `__del__` for the objects that are still alive.
- If an error happens inside `__del__`, Python doesn't stop the program: it only prints a warning.

So don't put important work (for example, saving data to a file) into `__del__`. It is better to write a normal method, for example `close()`, and call it yourself (or use `with`, as you do with files).

## The finalizer

A **finalizer** is a function that "finishes" an object: it runs when the object is being deleted and does the last cleanup work — for example, it closes a file or a network connection, or deletes a temporary file.

In Python, there are two kinds of finalizers:

1. **The method `__del__`.** The official Python documentation calls `__del__` a *finalizer*, and many people call it a *destructor*. In Python, these two words mean the same method. In C++, a destructor really destroys the object and frees its memory. In Python, the garbage collector frees the memory, and `__del__` only does the "last work" before that. This is why the word *finalizer* is more exact.
2. **The function `weakref.finalize`** from the standard library. It registers a function that Python will call when the object is deleted. This is a more reliable way to clean up:
   - the finalizer is called **only once**, even if we try to call it again;
   - it is also called when the program ends (if the object is still alive);
   - we can call it ourselves earlier, and we can check if it is still "alive".

The syntax is `weakref.finalize(obj, func, *args)`: when `obj` is deleted, Python calls `func(*args)`.

**Important:** don't pass the object itself (or its methods, like `self.close`) to `func` or `args`. If you do, the finalizer keeps a reference to the object, and the object will never be deleted.

```python
import weakref

def clean_up(name):
    print('Finalizer: cleaning up after', name)


class TempFile:
    def __init__(self, name):
        self.name = name
        print('TempFile', self.name, 'is created')
        # register the finalizer: clean_up(name) will be called when the object is deleted
        self._finalizer = weakref.finalize(self, clean_up, name)

    def close(self):
        # we can call the finalizer ourselves; the next calls will do nothing
        self._finalizer()

    def is_closed(self):
        return not self._finalizer.alive
```

```python
f = TempFile('data.tmp')
del f  # the last reference is deleted, so Python calls the finalizer
```

```python
f = TempFile('report.tmp')
f.close()            # the finalizer runs now
print(f.is_closed())
del f                # the finalizer is not called again
```

**Summary: the life of an object**

| Step | Method | Name |
|---|---|---|
| 1. The object is created | `__new__(cls, ...)` | constructor |
| 2. The object gets its attributes | `__init__(self, ...)` | initializer |
| 3. We work with the object | normal methods | — |
| 4. The object has no references and is deleted | `__del__(self)` / `weakref.finalize` | destructor / finalizer |

## Class attributes and instance attributes

Let's agree: an **attribute** of a class or an object is any element of the class/object (a variable, a method, a nested class) that we can get with a dot: `MyClass.<attribute>` or `my_object.<attribute>`.

The variables of a class are also called **fields** or **properties**. We can divide them into two groups:
- **static fields** — class attributes (class variables);
- **dynamic fields** — instance attributes (object variables).

What is the difference?

### 1. Static fields (class attributes)

These are variables that we declare in the body of the class. They are created when the class is created: we created the class — the variable exists.

```python
class Phone:

    # Static fields (class variables)
    default_color = 'Grey'
    default_model = 'C385'

    def turn_on(self):
        pass

    def call(self):
        pass
```

This is the list of the user-defined attributes of the class after we create it:

```python
print([name for name in dir(Phone) if not name.startswith('__')])
```

We can read or change such a field using only the name of the class. We don't need to create an instance for this:

```python
print(Phone.default_color)

# Change the default phone color from grey to black
Phone.default_color = 'Black'
print(Phone.default_color)
```

### 2. Dynamic fields (instance attributes)

These are variables that are created at the level of an instance. No instance — no variables. To create a dynamic field, we use `self` inside a method (usually inside `__init__`):

```python
class Phone:

    # Static fields (class variables)
    default_color = 'Grey'
    default_model = 'C385'

    def __init__(self, color, model):
        # Dynamic fields (object variables)
        self.color = color
        self.model = model
```

```python
# Create an instance of the class Phone: a red phone, model 'I495'
my_phone_red = Phone('Red', 'I495')

# The user-defined attributes of the new instance:
print([name for name in dir(my_phone_red) if not name.startswith('__')])

# Read the static fields of the object
print(my_phone_red.default_color, my_phone_red.default_model)

# Read the dynamic fields of the object
print(my_phone_red.color, my_phone_red.model)
```

```python
# Create another instance of the class Phone: the same phone, but in a different color
my_phone_blue = Phone('Blue', 'I495')

# Read the dynamic fields of the object
print(my_phone_blue.color, my_phone_blue.model)
```

Static fields are **shared** by all instances. Dynamic fields belong to **one** instance. Let's check it:

```python
Phone.default_color = 'White'  # change the class attribute
print(my_phone_red.default_color, my_phone_blue.default_color)

my_phone_red.color = 'Pink'    # change the instance attribute of one object
print(my_phone_red.color, my_phone_blue.color)
```

**What is `self` in Python?**

The word `self` is a reference to the current instance of the class. Usually Python passes this reference as the first parameter of a method:

```python
class Apple:

    # Create an object with 12 apples in total
    def __init__(self):
        self.whole_amount = 12

    # Eat some apples of the current object
    def eat(self, number):
        self.whole_amount -= number


apple = Apple()
apple.eat(5)
print(apple.whole_amount)
```

Note that `self` is **not a reserved word**. It is just a convention: we call the first parameter of a method `self`, and it gets a reference to the object that the method was called for. If you want, you can give the first parameter a different name — Python will work, but other programmers will not like it :)

In other programming languages (for example, Java or C++), the keyword `this` is used for the same purpose.

Also note that an object has both static attributes (at the class level) and dynamic attributes (its own, at the object level).

**Summary:**
- To create a static variable, it is enough to declare the class. The variable is created directly in the body of the class.
- Dynamic variables are created only when we work with an instance of the class.
- To create an instance variable, write `self.<variable_name>` inside one of the methods.
- An instance of a class has both static fields (class level) and dynamic fields (instance level).
- Different objects of the class can have different values of dynamic variables (and usually they do).

## Types of methods

As you already know, functions inside a class are called **methods**. There are three kinds of methods:
- instance methods (normal methods);
- static methods;
- class methods.

Let's see the difference.

### 1. Instance methods (normal methods)

You can use these methods only after you create an instance of the class: to call such a method, we call it on an instance. So the first parameter of such a method is `self`. As we discussed above, it gets a reference to the object that the method was called for. Here is an example:

```python
class Phone:

    def __init__(self, color, model):
        self.color = color
        self.model = model

    # A normal method
    # The first parameter of the method is self
    def check_sim(self, mobile_operator):
        if self.model == 'I785' and mobile_operator == 'MTS':
            print('Your mobile operator is MTS')
```

```python
# Create an instance of the class
my_phone = Phone('red', 'I785')

# Call the method check_sim() on the object my_phone
my_phone.check_sim('MTS')
```

Usually this is the biggest and the most used group of methods (compared to static methods and class methods).

### 2. Static methods

Static methods are normal functions that we put inside a class for convenience, so they live in the scope of this class. Usually it is some helper code.

The important thing is that you can call these methods using the class name, and you don't need to create an object. That's why these methods don't use `self` — they don't need any information about the objects of the class.

Remember the method `method_without_self` from the class `MyLittleClass4`? It worked only when we called it on the class. The right way to write such methods is to use the special decorator `@staticmethod`. Then the method works both on the class and on an instance:

```python
class Phone:

    # A static method for reference
    # It returns a hash by the model number
    # There is no self in the method
    @staticmethod
    def model_hash(model):
        if model == 'I785':
            return 34565
        elif model == 'K498':
            return 45567
        else:
            return None

    # A normal method
    def check_sim(self, mobile_operator):
        pass
```

```python
# Call the static method model_hash using only the class name
# We don't need to create a Phone object
print(Phone.model_hash('I785'))

# We can call it on an instance too
print(Phone().model_hash('K498'))
```

### 3. Class methods

Class methods are something in between normal methods (they are bound to an object) and static methods (they are bound only to the scope of the class). As you can guess from the name, these methods are closely connected with the class where they are defined.

Note that these methods can change the state of the class itself, and this affects ALL instances of the class. But they can't change one specific object of the class (instance methods do this).

To create a class method, use the decorator `@classmethod`. The first parameter of such a method is `cls`. Unlike `self`, `cls` is a reference to the class itself (not to an object). Let's look at an example:

```python
class Phone:

    def __init__(self, color, model, os):
        self.color = color
        self.model = model
        self.os = os

    # A class method
    # It takes 1) a reference to the class Phone and 2) a color as parameters
    # It creates a special Phone object (the special thing is that it is a toy phone)
    # To do this, it calls the initializer of the class Phone
    # and passes the color and the model of the toy phone to it
    @classmethod
    def toy_phone(cls, color):
        toy_phone = cls(color, 'ToyPhone', None)
        return toy_phone

    # A static method
    @staticmethod
    def model_hash(model):
        pass

    # A normal method
    def check_sim(self, mobile_operator):
        pass
```

```python
# Create a toy phone object
# Call the class method toy_phone using the class name and a dot
my_toy_phone = Phone.toy_phone('Red')
print(my_toy_phone.color, my_toy_phone.model, my_toy_phone.os)
```

And here a class method changes the state of the class, so ALL instances see the change:

```python
class Phone:
    default_color = 'Grey'

    def __init__(self, model):
        self.model = model

    # A class method: it changes the class attribute for all instances
    @classmethod
    def set_default_color(cls, color):
        cls.default_color = color


phone1 = Phone('I495')
phone2 = Phone('C385')
Phone.set_default_color('Black')
print(phone1.default_color, phone2.default_color)
```

As you can see from the examples, we often use class methods when:
- we need to create a special object of the current class;
- we need to use the factory pattern — create objects of different child classes inside the method.

**Summary:**

| Type of method | Decorator | First parameter | How to call it |
|---|---|---|---|
| instance method | — | `self` (the object) | `obj.method()` |
| static method | `@staticmethod` | — | `Class.method()` or `obj.method()` |
| class method | `@classmethod` | `cls` (the class) | `Class.method()` or `obj.method()` |

### Built-in and user-defined attributes

All attributes of a class can be divided into 2 groups:
- built-in (service) attributes;
- user-defined attributes.

**1. Built-in attributes.** The name "built-in" is a bit of a convention, and here is why. All classes in Python 3 have one common parent class — `object`. When you create a new class, it implicitly inherits from `object` and gets its attributes. These are the attributes we call built-in (service). Their names start and end with two underscores, for example `__new__`, `__init__` and `__del__`, which we discussed above. There are both fields and methods among them.

**2. User-defined attributes.** These attributes make the main functionality of the class. Built-in attributes come from the base class `object`, but the programmer writes user-defined attributes when they create the class and work with it.

Let's look at the full list of attributes of the simplest class:

```python
class Phone:
    pass

print(dir(Phone))
```

There are only built-in attributes here: our class inherited them from the base class `object` by default. Now let's add some functionality to it:

```python
class Phone:

    color = 'Grey'

    def turn_on(self):
        pass

    def call(self):
        pass

print(dir(Phone))
```

It is easy to see that three new user-defined attributes appeared in the list: the variable `color` and the methods `turn_on()` and `call()`.

**Summary:**
- Attributes are all the fields and methods of a class / object.
- Attributes are divided into built-in and user-defined ones.
- All classes in Python have a common parent class — `object`.
- The class `object` gives all its children a set of service attributes (both variables and methods).
- You can override many service attributes in your own class (as we did with `__init__`, `__new__` and `__del__`).
- The fields and methods that the programmer writes in the body of the class are user-defined. They are added to the list of attributes together with the built-in ones.

## Attribute privacy

In many programming languages, you can control access to the attributes of a class from the outside code and from other classes. Usually there are three access levels:
- public - the attributes are available both inside the class and from the outside code
- protected - the attributes are available inside the class and in the child classes
- private - the attributes are available only inside the class

Does Python have attribute privacy? Can we stop the outside code from reading and changing the attributes of an object?

```python
class VeryPrivateDataHolder:
    not_secret = 0     # public
    _secret = 1        # protected
    __very_secret = 2  # private
```

```python
obj = VeryPrivateDataHolder()
```

```python
obj.not_secret
```

```python
obj._secret
```

```python
obj.__very_secret
```

It looks like Python does have privacy after all, but there is a catch:

```python
obj._VeryPrivateDataHolder__very_secret
```

So if we really want to, we can still get access to "private" attributes. But we don't recommend it, especially with classes that are not yours!

```python
obj._VeryPrivateDataHolder__very_secret = 'new secret'
obj._VeryPrivateDataHolder__very_secret
```

Diamond inheritance is possible, but please don't do it!

```python
#      Animal
#    /       \
#  Cat       Dog
#    \       /
#      CatDog
```

**Diamond inheritance** is a situation in object-oriented languages with multiple inheritance: two classes B and C inherit from A, and class D inherits from both B and C. This scheme can be ambiguous. Imagine that an object of class D calls a method that is defined in class A (and class D did not override it), and classes B and C overrode this method in their own ways. Which class should D take the method from: B or C?

In Python, the diamond problem became serious after all classes got the common parent `object`. Since version 2.3, Python builds a resolution order list (MRO, Method Resolution Order).


```python
#      A
#   /    \
#  B      C
#   \   /
#     D
```

For the diamond, this means a depth-first search starting from the left $(D, B, A, C, A)$. Then Python removes from the list every repeated class, except its last occurrence. So the final resolution order is $[D, B, C, A]$.

As a result, we move layer by layer: we never go to a parent class before we visit all its children, even if this parent has several children.

```python
class CatDog(Cat, Dog):
    def __init__(self):
        super().__init__()
        print("i am a CatDog!")
```

```python
catdog = CatDog()
catdog.some_value
```

The order of the parents matters!

```python
class DogCat(Dog, Cat):  # now the other way round
    def __init__(self):
        super().__init__()#The super() function returns a proxy object
        #that passes method calls to a parent or sibling class of the given type
        print("i am a CatDog!")

dogcat = DogCat()
dogcat.some_value
```

What about methods?

Methods and attributes are searched in this order:

1) the name is searched in the object (that is, in its `__dict__`)

2) then in the class of the object

3) then in the parents of the class

```python
CatDog.__mro__
#The .__mro__ attribute contains the classes in exactly the order
#in which Python will search for methods in the class hierarchy
#until it finds the right one or raises an error.
```

```python
DogCat.__mro__
```

Multiple inheritance done right: mixins.

A mixin class is a programming idea: a class gives some functionality, but it is not meant to be used on its own. The main goal of mixins is to give some extra methods.

In other words, mixin classes (also called "mix-ins") are a limited form of multiple inheritance. In Python, a mixin is a parent class that gives some functionality to its child classes, but we don't create instances of the mixin itself. It is also better if mixins don't inherit from other mixins and don't have any state.

```python
class MusicPlayerMixin:
     def play_music(self, song):
        print("Now playing: {}".format(song))
```

```python
class Auto:
    def ride(self):
        print("Riding on a ground")

class Boat:
    def swim(self):
        print("Sailing in the ocean")

class Amphibian(Auto, Boat,MusicPlayerMixin):
    def __init__(self):
        super().__init__()
```

```python
a = Amphibian()
a.ride()
a.swim()
a.play_music('hi')
```

```python
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds"
        self.balance -= amount
        return "The amount was withdrawn"

account = BankAccount(1000)
print(account.withdraw(300))
print(account.balance)
```
