# Part 1. `list`

## 1.1 Containers

A **container** is a data type whose values contain other values. One container
object holds many other objects inside it.

Python has four built-in containers:

| Type | Name | Written as | Ordered | Mutable |
|---|---|---|---|---|
| `list` | list | `[1, 2, 3]` | yes | yes |
| `tuple` | tuple | `(1, 2, 3)` | yes | **no** |
| `dict` | dictionary | `{'a': 1}` | by insertion | yes |
| `set` | set | `{1, 2, 3}` | **no** | yes |



`list` and `tuple` are **sequences**: their items are kept in a fixed order, and
we can take any item by its **index** (its position number). A `set` is not a
sequence — it has no order and no indices at all.

> If you know C++: a Python `list` is close to `std::vector`, and a Python
> `tuple` is close to a `const` array. But there is one big difference — a
> Python container may hold items of **different** types at the same time,
> including other containers.

## 1.2 Creating a list

An empty list can be written in two ways: with empty square brackets `[]`, or
with the function `list()`.

````python
a = []
b = list()

# let us check that the two ways give the same result:
print(a == b)
print(type(a), type(b))
print(len(a))
````

A list with items is written as a comma-separated sequence inside square
brackets. The items may have **different** types — numbers, strings, `None`,
`bool`, and even other lists:

````python
list_with_diff_basic_types = [1, 2.0, 'string!', [None], False]
print(list_with_diff_basic_types)
print(len(list_with_diff_basic_types))
````

Usually, however, a list holds items of one kind, because we want to do the same
thing with all of them. Here is the example we will use through the whole
lesson — a wish list:

````python
wish_list = ['smart-watch', 'notebook', 'iphone']
print(wish_list)
````

The function `list()` can also build a list from any object we can loop over
(any **iterable** — see lesson 2). This is a very common way to get a list:

````python
print(list('hello'))        # a string -> a list of its characters
print(list(range(5)))       # a range  -> a list of numbers
print(list((1, 2, 3)))      # a tuple  -> a list
````

## 1.3 Indexing

Each item of a list has an **index** — its position. Python counts positions
**from zero**, so the first item has index `0`, the second one index `1`, and so
on.

| item | `'smart-watch'` | `'notebook'` | `'iphone'` |
|---|---|---|---|
| **index** | 0 | 1 | 2 |
| **negative index** | -3 | -2 | -1 |

````python
print(wish_list[0])
print(wish_list[1])
````

Python indices can be **negative**. A negative index counts from the end, so
`-1` is the last item, `-2` is the item before the last one, and so on. This is
much shorter than writing `wish_list[len(wish_list) - 1]`.

````python
print('List:              ', wish_list)
print('Last item:         ', wish_list[-1])
print('The item before it:', wish_list[-2])
````

If there is no item with the index we ask for, Python stops the program with an
**`IndexError`**:

````python
wish_list[10]
````

An index must be an **integer**. A `float` does not work, even when it looks
like a whole number, because Python does not guess what we mean:

````python
wish_list[1.0]
````

An index does not have to be a literal number. We can (and very often must) use
a **variable**, and we can also write an arithmetic expression or call a
function inside the square brackets:

````python
index = 0
print(wish_list[index])
print(wish_list[2 * index + int(1.0)])   # 2 * 0 + 1 -> index 1
print(wish_list[len(wish_list) - 1])     # the last item, the long way
````

## 1.4 Changing a list

A list is **mutable**: we can add, remove and replace its items, and the object
stays the same object.

To replace one item, we assign a new value to its index:

````python
print('Before:', wish_list)

wish_list[0] = 'smart-watch xiaomi s1'   # a more exact model
print('After: ', wish_list)
````

To add items, we have three methods:

* `append(x)` — add **one** item to the end;
* `insert(i, x)` — put an item **before** position `i`, and move the rest to the
  right;
* `extend(other)` — add **all** items of another container to the end.

````python
wish_list.append('car')          # we also want a car
print(wish_list)

wish_list.insert(1, 'dog')       # a dog, and it should be second in the list
print(wish_list)
````

`append` and `extend` are easy to confuse. `append` always adds **one** item,
even when that item is itself a list; `extend` adds the items **one by one**:

````python
a = [1, 2, 3]
b = [1, 2, 3]

a.append([4, 5])     # one new item, and that item is a list
b.extend([4, 5])     # two new items

print(a)
print(b)
````

To remove items, we have three ways:

* `remove(x)` — remove the **first** item that is equal to `x`
  (`ValueError` if there is no such item);
* `pop()` — remove **and return** the last item; `pop(i)` does the same for
  position `i`;
* `del my_list[i]` — remove the item at position `i` and return nothing.

````python
wish_list.remove('iphone')       # we have already bought it
print(wish_list)
````

`pop` is different from the other two, because it **gives the item back** to us.
This is useful when we want to use the item after removing it:

````python
last = wish_list.pop()
print('The list without the last item:', wish_list)
print('The item we took:              ', last)

wish_list.append(last)           # put it back, we still want it
print('The list again:               ', wish_list)
````

With an argument, `pop(i)` removes the item at position `i`:

````python
second = wish_list.pop(1)
print(wish_list)
print(second)

wish_list.insert(1, second)      # put it back in its place
print(wish_list)
````

Here are the methods of a list in one table. `n` is the length of the list.

| Method | What it does | Returns |
|---|---|---|
| `lst.append(x)` | adds one item to the end | `None` |
| `lst.insert(i, x)` | puts `x` before position `i` | `None` |
| `lst.extend(other)` | adds all items of `other` | `None` |
| `lst.remove(x)` | removes the first item equal to `x` | `None` |
| `lst.pop()` / `lst.pop(i)` | removes the last item / the item `i` | **the item** |
| `lst.clear()` | removes everything | `None` |
| `lst.index(x)` | position of the first item equal to `x` | `int` |
| `lst.count(x)` | how many items are equal to `x` | `int` |
| `lst.sort()` | sorts the list itself | `None` |
| `lst.reverse()` | turns the list around | `None` |
| `lst.copy()` | a new list with the same items | **a list** |

> **A very common mistake.** Almost all these methods change the list **in
> place** and return `None`. If we write
>
> ```python
> wish_list = wish_list.append('car')
> ```
>
> the list is changed, but then the name `wish_list` is attached to `None`, and
> our data is lost. Write `wish_list.append('car')` on its own line, with no
> assignment.

````python
demo = [1, 2, 3]
result = demo.append(4)

print(demo)      # the list did change
print(result)    # ...but the method returned None
````

## 1.5 Arithmetic with lists

Two lists can be **added**. The result is a **new** list: first all the items of
the left list, then all the items of the right one.

````python
a = [1, 2, 3]
b = [4, 5, 6]

print(a + b)
print(a)         # a itself did not change
````

Lists cannot be subtracted — there is no meaning for "list minus list" in
Python, so we get a `TypeError`:

````python
print(a - b)
````

Like strings, a list can be **multiplied by an integer**. The result is the list
repeated several times:

````python
print([2, 3, 4] * 3)
print([0] * 5)           # a popular way to build a list of five zeros
````

> **A trap with `*` and nested lists.** Multiplication copies the **names**, not
> the objects — exactly as in lesson 2. So `[[0] * 3] * 3` gives us three names
> for **one** inner list, and changing one "row" changes all of them:

````python
bad_matrix = [[0] * 3] * 3
bad_matrix[0][0] = 1
print('bad: ', bad_matrix)      # all three rows changed!

good_matrix = [[0] * 3 for _ in range(3)]   # a comprehension builds a new row each time
good_matrix[0][0] = 1
print('good:', good_matrix)
````

Finally, `+=` and `+` are **not** the same for lists. `a = a + b` builds a new
list and moves the name to it; `a += b` changes the old list itself, so every
other name that points to it sees the change.

````python
a = [1, 2]
copy_of_name = a
a += [3]                 # in place: the same object grows
print(a, copy_of_name)

a = [1, 2]
copy_of_name = a
a = a + [3]              # a new object; copy_of_name still sees the old one
print(a, copy_of_name)
````

## 1.6 Is the item in the list?

The operator `in` answers this question and gives us `True` or `False`:

````python
print(wish_list)
print('iphone' in wish_list)
print('car' in wish_list)
print('car' not in wish_list)
````

> **A note about speed.** To answer `in`, Python compares the item with the
> items of the list one by one, so the cost grows with the length of the list
> (`O(n)`). For a `set` or a `dict` the same question costs `O(1)` — see lesson
> 2. If your program asks "is it there?" very often, a `set` is the better tool.

## 1.7 Slices

A **slice** takes a whole *range* of items at once. It works for every sequence:
strings, lists and tuples. The syntax is

```python
variable[start:stop:step]
```

where `start` is the first index (**included**), `stop` is the index where we
stop (**excluded**), and `step` is the size of the step. Any of the three may be
left out. The default values are `start = 0`, `stop = len(...)`, `step = 1`.

> **Note**
>
> Slices work for all containers whose items can be reached by index. We will
> meet them again when we work with libraries — for example with NumPy, where
> slices are one of the main tools.

Let us look at an example on a string:

````python
string = 'abcdefgh'
print(string[1:7:2])
````

We took the items from index `1` (`'b'`) up to index `7` (`'h'`, **not**
included), with step 2. The same syntax works for lists and tuples:

````python
list_integers_range = [1, 2, 3, 4, 5]
print(list_integers_range[::2])      # every second item: indices 0, 2, 4
````

The step may be **negative**, and this is the shortest way to turn a sequence
around:

````python
print(list_integers_range[::-1])
print('abcdefgh'[::-1])
````

When we leave a part of the slice out, Python uses the default value. Read the
comments in the cell below — this is the whole idea of slices in five lines:

````python
print(list_integers_range)
print(list_integers_range[1:4:2])  # from index 1 to index 4 (excluded), step 2
print(list_integers_range[1:4])    # the same, with the default step 1
print(list_integers_range[:4])     # the default start is 0
print(list_integers_range[1:])     # the default stop is the end of the list
print(list_integers_range[:])      # everything -- a full copy
````

Negative indices are allowed inside a slice too, and they mean the same as
before — positions counted from the end:

````python
letters = ['a', 'b', 'c', 'd', 'e', 'f']
print(letters[-3:])      # the last three items
print(letters[:-3])      # everything except the last three
print(letters[-3:-1])    # from the third from the end to the last (excluded)
````

> **A useful difference.** An index that is too large gives an `IndexError`, but
> a **slice** that is too large gives no error at all — Python simply returns
> what exists:

````python
print(letters[100:200])   # an empty list, not an error
print(letters[3:100])     # everything that exists from index 3
````

A slice can also stand on the **left** side of an assignment. Then it replaces
the whole range, and the new part does **not** need to have the same length. In
the example below two numbers are replaced by three:

````python
list_integers_range = [1, 2, 3, 4, 5]
print(list_integers_range)

list_integers_range[1:2] = [200, 300, 400]
print(list_integers_range)

list_integers_range[1:4] = []          # an empty right side deletes the range
print(list_integers_range)
````

## 1.8 Finding an item

Indexing answers the question "which item is at position 3?". Sometimes we need
the opposite: "at which position is this item?". The method `index` does that.

`lst.index(x)` returns the position of the **first** item equal to `x`. If there
is no such item, we get a `ValueError`:

````python
numbers = [1, 200, 300, 400, 3, 4, 5, 400]

print(numbers.index(400))
print(numbers.count(200))
print(numbers.index(10))     # 10 is not in the list -> ValueError
````

Because `index` raises an error, it is often used together with `in`:

````python
if 10 in numbers:
    print(numbers.index(10))
else:
    print('there is no 10 in the list')
````

Python also has short built-in functions for the usual questions about a list of
numbers:

````python
print(min(numbers), max(numbers), sum(numbers), len(numbers))
print(sum(numbers) / len(numbers))     # the average value
````

## 1.9 Sorting a list

There are two ways to sort, and the difference between them is important.

* `sorted(lst)` — a **function**. It returns a **new**, sorted list and leaves
  the old one as it was.
* `lst.sort()` — a **method**. It sorts the list **itself** and returns `None`.

````python
numbers = [1, 200, 300, 400, 3, 4, 5]

numbers_sorted = sorted(numbers)
print('the original list:', numbers)
print('the new list:     ', numbers_sorted)
````

````python
numbers = [1, 200, 300, 400, 3, 4, 5]

sort_return = numbers.sort()
print('the list itself:  ', numbers)
print('what sort returned:', sort_return)
````

Both accept the argument `reverse=True`, which sorts from big to small, and the
argument `key=`, which says *by what* we sort. `key` takes a function, and
Python sorts by the result of that function:

````python
words = ['banana', 'fig', 'apple', 'kiwi']

print(sorted(words))                    # alphabetical order
print(sorted(words, reverse=True))      # backwards
print(sorted(words, key=len))           # by the length of the word
````

Lists of strings are sorted in **lexicographic** order — like words in a
dictionary, character by character. Careful: digits inside a string are compared
as characters, so `'42'` comes before `'8'`, and capital letters come before
small ones:

````python
list_strings = ['9', '4', '42', '8', 'a', 'bb', 'bac', 'Zebra']
print(sorted(list_strings))
````

And what about a list with items of **different** types? Python refuses: it does
not know whether a number is smaller or bigger than a list, so we get a
`TypeError`:

````python
list_diff_types = [[1, 2], 1, 2]
print(sorted(list_diff_types))
````

## 1.10 Copying a list

This is the place where beginners lose the most time, so we go slowly. Remember
the rule from lesson 2: **assignment never copies an object, it only adds a
second name.**

````python
list1 = [1, 2, 3]
list2 = list1          # NOT a copy: one list, two names

list2.append(4)
print(list1)           # list1 changed too
print(id(list1) == id(list2))
````

If we want an independent list, we must ask for a copy. There are three usual
ways, and they all do the same thing:

````python
list1 = [1, 2, 3]

list2 = list1.copy()   # a method
list3 = list1[:]       # a full slice
list4 = list(list1)    # the function list()

list2.append(4)
print(list1, list2, list3, list4)
````

> **Careful: the copy is "shallow".** A copy makes a new outer list, but the
> items inside are the **same objects**. If an item is itself a list, both
> copies share it:

````python
import copy

original = [[1, 2], [3, 4]]

shallow = original.copy()
shallow[0].append(99)
print('after a shallow copy: ', original)      # the inner list changed here too

original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)                 # copies everything, on all levels
deep[0].append(99)
print('after a deep copy:    ', original)
````

| What we write | What we get |
|---|---|
| `b = a` | one list, two names |
| `b = a.copy()`, `b = a[:]`, `b = list(a)` | a new outer list, the same items inside |
| `b = copy.deepcopy(a)` | new objects on every level |

## 1.11 Lists and strings: `split` and `join`

Two string methods build a bridge between text and lists. They appear in almost
every real program, so it is worth knowing them now.

* `text.split(sep)` — cuts a string into a **list** of parts;
* `sep.join(lst)` — glues a list of **strings** back into one string.

````python
line = 'ivan peter maria'
parts = line.split()             # by spaces, when we give no argument
print(parts)

print('19,7,42'.split(','))      # by a comma

print(', '.join(parts))          # back to a string
print('-'.join(['2026', '09', '07']))
````

# Part 2. `tuple`

## 2.1 What a tuple is

A **tuple** is a container whose items are reached by index, exactly like in a
list, but which **cannot be changed** after it is created.

**TL;DR: a tuple is an immutable list.**

We cannot add, remove or replace its items. Everything that only *reads* the
tuple still works.

## 2.2 Creating a tuple

The usual way is round brackets:

````python
tuple_diff_types_example = (1, 2.0, '3', (4,), [None])

print(tuple_diff_types_example)
print(type(tuple_diff_types_example))
````

> **Careful!** To create a tuple of **one** item, we must put a comma after it.
> Without the comma, the brackets are just ordinary brackets around a value:

````python
tuple_1 = (4)      # this is the number 4 in brackets
tuple_2 = (4,)     # this is a tuple with one item

print(type(tuple_1))
print(type(tuple_2))
````

In fact, it is the **comma** that makes a tuple, not the brackets. We can write
a tuple without any brackets at all — but this sometimes makes the code harder
to read, so use it only when the meaning is clear:

````python
tuple_example = 'a', 5, 12.345, (2, 'b')
print(tuple_example)
print(type(tuple_example))
````

An empty tuple is the one case with no comma: `()` or `tuple()`. And, like
`list()`, the function `tuple()` builds a tuple from any iterable:

````python
print(tuple(), type(tuple()))
print(tuple([1, 2, 3]))       # from a list
print(tuple('abc'))           # from a string
print(tuple(range(4)))        # from a range
````

## 2.3 A tuple cannot be changed

Let us check it. A tuple has no method `append`, and item assignment is not
allowed either:

````python
print(tuple_example)
tuple_example.append(5)
````

````python
tuple_example[0] = 9
````

The error messages are worth reading. The first one says `AttributeError`:
there is simply no such method. The second one says
`'tuple' object does not support item assignment` — the operation exists in the
language, but not for this type.

We can still build a **new** tuple from the old one, and this is what `+` does:

````python
m = (1, 2, 3)
print(tuple_example)
print(tuple_example + m)      # a new tuple; tuple_example itself is untouched
print(tuple_example)
````

> **The famous trap.** A tuple protects **its own** items — that is, the
> references it holds. It does not freeze the objects those references point to.
> If we put a **list** inside a tuple, that list is still mutable:

````python
t = ([1, 2], 5)
print(t)

t[0].append(4)      # we do not change the tuple: we change the list inside it
print(t)
````

Nothing was broken here. `t[0]` is the same list object as before — its length
changed, but the tuple still points to the same two objects. This is also the
answer to the first task in Part 3, and it has one important result:

**such a tuple is not hashable.** In lesson 2 we learned that a `dict` key must
be immutable. A tuple of numbers and strings can be a key; a tuple with a list
inside cannot:

````python
print(hash((1, 2, 3)))           # fine
print(hash(([1, 2], 5)))         # TypeError: unhashable type: 'list'
````

## 2.4 What we can do with a tuple

Everything that does not change it. Indexing, slices, `len`, `in`, `+`, `*`,
`count`, `index`, `min`, `max`, `sorted` — all of these work:

````python
tuple_example = ('a', 5, 12.345, (2, 'b'))

print(tuple_example[2])          # indexing
print(tuple_example[:2])         # a slice -- and a slice of a tuple is a tuple
print(tuple_example.index(5))    # searching
print(len(tuple_example))        # the size
print(5 in tuple_example)        # is the item there?
print(tuple_example * 2)         # repetition
````

Note that `sorted()` always returns a **list**, even when we give it a tuple —
it cannot return a tuple, because it builds the result step by step:

````python
t = (5, 1, 4, 2)
print(sorted(t))              # a list
print(tuple(sorted(t)))       # ...and back to a tuple, if we need one
````

## 2.5 Unpacking

This is the feature that makes tuples really pleasant to use. Python can take a
tuple apart and put its items into several variables at once:

````python
a, b = 1, 2      # the right side is the tuple (1, 2)
print(a, b)

a, b = b, a      # the famous Python swap: no temporary variable needed
print(a, b)
````

The version with brackets means exactly the same thing:

````python
(a, b) = (1, 2)
print(a)
print(b)
````

The number of names on the left must be equal to the number of items on the
right, otherwise we get a `ValueError`:

````python
x, y = (1, 2, 3)
````

If we do want "the rest", we mark one name with a star. It collects everything
that is left, always as a **list**:

````python
first, *rest = [10, 20, 30, 40]
print(first, rest)

first, *middle, last = [10, 20, 30, 40]
print(first, middle, last)
````

Unpacking also happens inside `for` loops — every time we write `for key, value
in ...`, Python is unpacking a tuple for us. Both `enumerate` and `zip` from
lesson 2 give us tuples:

````python
wish_list = ['smart-watch', 'notebook', 'iphone']
prices = [200, 1000, 900]

for i, item in enumerate(wish_list):
    print(i, item)

print()
for item, price in zip(wish_list, prices):
    print(item, '->', price)
````

> When one of the values is not interesting, the usual name for it is `_`
> (an underscore). It is an ordinary variable name, but every Python programmer
> reads it as "I do not need this one".

````python
point = (3, 7, 'red')
x, y, _ = point
print(x, y)
````

## 2.6 Why do we need a tuple at all?

**The question.** Why use a `tuple` if a `list` can do more?

**The answers.**

1. **A tuple is faster and smaller.** Python knows its size will never change,
   so it can store it more efficiently. For a constant collection of values a
   tuple is the better choice.
2. **A tuple protects the data.** If a value must not change, a tuple turns a
   possible bug into an immediate error. This is documentation that the computer
   checks for us.
3. **A tuple can be a `dict` key or a `set` item**, because it is hashable — as
   long as all its items are hashable. A list can never do this.
4. **A tuple is the natural way to return several values** from one function.

````python
import sys

print('list: ', sys.getsizeof([1, 2, 3, 4, 5]), 'bytes')
print('tuple:', sys.getsizeof((1, 2, 3, 4, 5)), 'bytes')
````

Point 3 is the one we meet most often in real code. A pair of coordinates, a
(day, month) pair, a (name, surname) pair — all of these are perfect `dict`
keys:

````python
board = {}
board[(0, 0)] = 'rook'
board[(0, 4)] = 'king'

print(board)
print(board[(0, 4)])
print((0, 1) in board)
````

And point 4 — a function that returns two values in fact returns one tuple, and
we unpack it at once. The built-in function `divmod` works exactly like this:

````python
result = divmod(17, 5)
print(result, type(result))

quotient, remainder = divmod(17, 5)
print(quotient, remainder)
````

## 2.7 `list` and `tuple` side by side

| | `list` | `tuple` |
|---|---|---|
| Written as | `[1, 2, 3]` | `(1, 2, 3)` |
| Ordered | yes | yes |
| Access by index and slices | yes | yes |
| Mutable | **yes** | **no** |
| `append`, `remove`, `sort` | yes | **no** |
| Hashable (a `dict` key) | **no** | yes, if all its items are |
| Memory and speed | more | less |
| Typical use | data that grows and changes | a fixed record of values |

**A simple rule of thumb.** If the collection describes *many things of the same
kind* and will change — use a `list`. If it describes *one thing with several
parts* and must stay as it is — use a `tuple`.
